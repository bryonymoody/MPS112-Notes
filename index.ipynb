# Introduction to the module
 
Welcome to the webpages and learning materials for MAS112.

Below are some key details about the module:

* Lecture notes
These lecture notes detail all the mathematical knowledge you need for this semester of MPS112. Please have a copy with you when you watch the lectures as I will refer to them in lectures.

Any non-examinable content will be clearly highlighted throughout the notes, all other content can be assumed to be non-examinable.


* Blackboard
This is the online repository for all information about the module. Please check this regularly (you will also get email updates).
* Obtaining help with the module
Please use the Dicussion Board on Blackboard to ask questions in the first instance - I will check this regularly. You are also strongly encouraged to come along to my student drop-in hours for in-person help.
* Homework and practice questions
At the end of each topic, there are some exercises. We may cover a small number of these in the lectures, but any we do not cover, you should do as independent study.
* Tutorials
There are no formal tutorials for mathematical content of MPS112, instead each lecture will consist of spending some time learning new content, and the rest of the time practicing using exercises.
* Assessment
You will be assessed with regular tests, the information about these is available in the assessment section on Blackboard.
* Lectures
You are encouraged to engage in **active learning**. This means answering questions in lectures when able, and having a go at questions before looking at the answers.
