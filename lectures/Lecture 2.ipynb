# Topic 2: Straight line graphs

## Chemistry motivation
As chemists, you will find your self working with data, often needing to determine the relationship between variables or analyse trends over time. Graphs allow us to do this is a visual way, but they are easy to get wrong! This section will revise straight line graphs and give you some tips for plotting graphs that will help you, and others, get the most out of your experimental data. 

## Mathematical theory

Representing data on a graph allows us to easily identify trends and patterns that are present. A graph can be used to show the relationship between two related values, the **dependent**
and the **independent variables**. An independent variable (or a control variable} is one that we can change. It is often represented by the algebraic symbol $x$. Whereas the dependent variable (or the observed variable) is the one that we measure and whose value depends on the independent variable. It is often represented by the algebraic symbol $y$.

![](straigh_line.png)

Always plot the independent or control variable along the horizontal (or $x$) axis and the dependent or observed variable along the vertical (or $y$) axis. 

Straight line graphs come in the form:
\begin{equation*} y = \textcolor{blue}{m}x + \textcolor{red}{c}  \end{equation*}
where $x$ is the independent variable, $y$ is the dependent variable. 

We then have $\textcolor{red}{c}$ is known as the **intercept** and this is where the graph passes through the $y$-axis. This tells us what value we expect the dependent variable to take when the independent value takes a value of 0.  

Finally, $\textcolor{blue}{m}$ is known at the **gradient or slope** and decsribes how steep the line it. It tells us how quickly the value of the dependent variable changes when we change the independent variable. In other words, it tells us that for every increase of 1 unit of the independent variable, we expect the independent variable to change by $m$ units. 

```{prf:axiom} Calculating the gradient of a straight line
To calcuate the gradient, you take two points $(x_1, y_1)$ and $(x_2, y_2)$ and find the change in the $y$ values and divide it by the change in $x$ values,
\begin{equation*} \text{Gradient} = \frac{y_2 - y_1}{x_2 - x_1} = \frac{y_1 - y_2}{x_1 - x_2} \end{equation*}
```
```{note} Notation
You may also see $y$ represented as $f(x)$, you should read $y = f(x)$ as "$y$ is equal to a function of $x$".

Similarly, you might see $y_2 - y_1$ denoted as $\Delta y$ and you should read this as "change in $y$"
```
### Different slopes

You can have horizontal lies with zero slope, these are lines of the form $y = c$ and tell us that the independent variable does not influence the dependent variable. For graphs of the form $y = mx + c$, if $m$ is positive then this tells us the line will slant upwards, running from bottom left to top right as in the graph above. However, $m$ can also be negative, which tells us that for every increase in 1 unit of the independent variable, the dependent variable decreases by $m$ units.  

```{note} Note
Graphs usually have physical significance and hence include units.
For example, the graph of a particle’s velocity as a function of time:
In this graph below, the equation of the line is $v = 8t + 25$ where $v$ has units of ms$^{-1}$

For the equation to work, everything on the right-hand side must also have units of ms$^{-1}$. Hence $c = 25$ has units of ms$^{-1}$. The starting velocity of the particle at $t = 0$ s is $25 \text{ms}^{-1}$. The units of m can be worked out from the equation: 
\begin{equation*}  m = \frac{\Delta y}{\Delta x} \text{ or in this case } \frac{\Delta v}{\Delta t} . \end{equation*}

We know that $ \Delta v$ has the same units as $v$ (ms$^{-1}$) and similarly $\Delta t$ has the units (s). Hence the units of the gradient, $m$ are
\begin{equation*} \frac{\text{ms}^{-1}}{\text{s}} = \text{ms}^{-1} \times \text{s}^{-1} = \text{ms}^{-2}. \end{equation*}
```
