# Experimental Errors, Accuracy and Precision

1. Define the following terms as used in experimental chemistry:
   a. Accuracy: How close a measured experimental value is to the true or accepted reference value. <br>
   b. Precision: How close repeated experimental measurements are to one another (reproducibility/repeatability), regardless of whether they match the true value. <br>

2. A series of four standard titrations were carried out to determine the concentration of a sample of hydrochloric acid. The recorded burette readings for the volume of $\text{NaOH}$ required were:
   * Trial 1: $24.80\text{ cm}^3$
   * Trial 2: $23.15\text{ cm}^3$
   * Trial 3: $23.20\text{ cm}^3$
   * Trial 4: $23.10\text{ cm}^3$
   
   The true reference value for the reaction volume is $23.18\text{ cm}^3$.<br>
   a.
    \begin{equation*} \\[1em] \text{Mean} = \frac{24.80 + 23.15 + 23.20 + 23.10}{4} = 23.56\text{ cm}^3.\end{equation*}  
   b. We observed high precision, as the readings are within $\pm 0.5\text{ cm}^3$ of each other. Similarly, we see high accuracy, because the mean titre ($23.56\text{ cm}^3$) is within $1.6\%$ of the true reference value ($23.18\text{ cm}^3$).<br>

3. $\text{Percentage Error} = \left(\frac{0.010\text{ g}}{0.250\text{ g}}\right) \times 100\% = 4.0\%$
