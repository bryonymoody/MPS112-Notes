## Mathematical theory

### Standard deviation and variance 
Recall that we have two standard deviations that we might encounter, 
```{prf:axiom} Population standard deviation
\begin{equation*}
\sigma = \sqrt{\frac{1}{N} \Sigma^N_{i = 1} (x_i - \mu)^2}
\end{equation*}

such that $N$ is the population size and $\mu$ is the population mean.

```

```{prf:axiom} Sample standard deviation
\begin{equation*}
s = \sqrt{\frac{1}{n-1} \Sigma^n_{i = 1} (x_i - \bar{x})^2}
\end{equation*}

such that $n$ is your sample size and $\bar{x}$ is the sample mean.
```
If we square these terms, we get the **populations variance**, $\sigma^2$ and the **sample variance**, $s^2$.

There is a different way to calcuate the variance which you might see defined, such that

\begin{equation*}
\sigma^2 = \frac{\sum^N_{i=1} x^2}{N} - \left(\frac{\sum^N_{i=1}x}{N}\right)^2 \newline \\[1em]
=\overline{x^2} - (\bar{x})^2
\end{equation*}

```{hint} But why? (Non-examinable)
:class: dropdown

Our initial defintion of variance we defined as 
\begin{equation*} \frac{1}{N} \sum^N_{i=1}(x_i-\bar{x})^2 \end{equation*}
which is simple our standard deviation, but we can expand the brackets giving

\begin{equation*} \frac{1}{N} \sum^N_{i=1}(x_i^2- 2x_i\bar{x} + \bar{x}^2) \end{equation*} \\[1em]
\begin{equation*} =  \frac{1}{N} \sum^N_{i=1} x_i^2 - 2\bar{x} \textcolor{red}{\frac{1}{N} \sum^N_{i=1} x_i} + (\bar{x})^2 \frac{1}{N}\textcolor{blue}{ \sum^N_{i=1} 1} \\[1em]

= \frac{1}{N} \sum^N_{i=1} x_i^2 - 2\bar{x} \times \textcolor{red}{\bar{x}} + (\bar{x})^2 \frac{1}{N}\times \textcolor{blue}{N},
\end{equation*}

which simplifies to 
\begin{equation*} =  \frac{1}{N} \sum^N_{i=1}x_i^2 - 2\bar{x}^2 + (\bar{x})^2 \\[1em]
= \overline{x^2} - (\bar{x})^2.
\end{equation*}



```
In other words, you simply square all your data points, find the average and then take away the sample mean squared. Either method of working out the standard deviation and variance is valid, and in face, in almost all cases you would be using a computer program such as Excel to carry out these calculations anyway. 

### Covariance and correlation
Variance allows us to look at the spread of the data for one variable, but we might have two variable and want to quantify how these two variables are chaging simultaneously. To do this, we are first going to define some additional summary statistics for two variables $x$ and $y$. 

```{prf:axiom} Mean of the product $xy$
\begin{equation*} 
\overline{xy} = \frac{1}{N}\sum^N_{i=1} x_iy_i
\end{equation*}
```
```{prf:axiom} Covariance

The **population covariance** of two variables $x$ and $y$  is defined as 
\begin{equation*}
cov(x,y) = \frac{1}{N}\sum^N_{i=1} (x_i - \mu_x)(y_i - \mu_y) 
\end{equation*}
which can also be rewritten as 
\begin{equation*}
cov(x,y) = \mu_{xy} - \mu_{x}\mu_{y}.
\end{equation*}

whereas the **sample covariance** is defined as
\begin{equation*}
cov(x,y) = \frac{1}{n-1}\sum^n_{i=1} (x_i - \bar{x})(y_i - \bar{y}) 
\end{equation*}
which can also be rewritten as 

\begin{equation*}
cov(x,y) = \frac{n}{n-1}\overline{xy} - \bar{x}\bar{y}
\end{equation*}

```

The **covariance** provides us with that measure of how two variables simultaneously. Similarly **correlation** measures this too, but is standardised between -1 and 1

```{prf:axiom} Correlation
The correlation coefficient for a given **sample**, often seen denoted as either $\rho_{x,y}$ or $cor(x,y)$ is defined as
\begin{equation*}
cor(x,y) = \frac{cov(x,y)}{s_x s_y} = \frac{\overline{xy} - \bar{x}\bar{y}}{s_x s_y} \times \frac{n}{n-1}
\end{equation*}
```

In the case where the values of $x$ and $y$ are random and uncorrelated (as is also the case for random experimental errors), the measured value of $x_i$ has no bearing on the measured value of$y_i$. Thus, the covariance term $cov(x,y) = 0$, and (subsequently the correlation term is also 0) for random and independent variables.

First, we will work out the covariance. Let $x$ be hours spent revising and $y$ be the test score of a student. From the data we find that $\bar{x} = 4$, $\bar{y} = 53.5$ and $\overline{xy} = 214$. Using 
\begin{equation*}
\\[1em]
cov(x, y) = \overline{xy} - (\bar{x}\bar{y}) \times \frac{n}{n-1}
\end{equation*}
since we are working with a sample, we find 
\begin{equation*}
\\[1em]
cov(x,y) = 267 - (4 \times 53.5) * \frac{6}{5} = 63.6. \\[1em]
\end{equation*}

Now we have found the covariance, we simply need to divide by the product of the sample standard deviations to find the correlation:
\begin{equation*}
cor(x,y) = \frac{cov(x,y)}{s_x s_y} = \frac{63.6}{2.61 \times 24.94} = 0.98,
\end{equation*}
which is a very strong correlation, which we suspected from our initial plot.
