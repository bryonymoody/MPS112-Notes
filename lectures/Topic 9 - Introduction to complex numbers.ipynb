# Topic 9: Introduction to complex numbers 

## Chemistry motivation
Complex numbers are not complicated or difficult, instead they are "complex" because they are composed of two distinct parts, which we will investigate in this topic. 

They are used in many areas of physics, chemistry, engineering:
- Wavefunctions (electronic structure of molecules) 
- X-ray scattering analysis (crystal structure of a solid) 
- Interaction of radiation with matter (optical properties) 
- Fluid dynamics (weather forecast) 
- Computer graphics animation (movies, games) 
- Signal processing (mobile phone) 

## Mathematical theory

As seen in GCSE, the standard form of a quadratic equation is:

\begin{equation*} ax^2 + bx + c = 0 \end{equation*}

where $a \neq 0$.

Its solutions are given by the quadratic formula:

\begin{equation*} x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}\end{equation*}

If $\sqrt{b^2 - 4ac}$ (known as the discriminant) is more than zero then this equation has two distinct solutions. If the discrimintant is equal to 0, then we have a single solution. 

However, we run into issues when $\sqrt{b^2 - 4ac}$ is less that 0. Since trying to find the square root of a negative number will lead to a math error in your calculator. This is where imaginary numbers come in. 

### The Imaginary Number $i$

```{prf:axiom} Imaginary unit
We define the imaginary unit $i$ such that: 
\begin{equation*} i = \sqrt{-1} \quad \text{or} \quad i^2 = -1\end{equation*}
```

Higher powers of $i$ follow a cyclical pattern of length 4:
- $i^1 = i$
- $i^2 = -1$
- $i^3 = i^2 \times $i = -i$
- $i^4 = (i^2)^2 = 1$


```{seealso} Example
Suppose we are trying to find the solution to the equation $x^2 + 1 = 0$. Rearraging we get 
\begin{equation*}x^2 = -1,\end{equation*}
therefore we find 
\begin{equation*}x = \pm \sqrt{-1} = \pm i\end{equation*}
```

## The General Form of Complex Numbers

```{prf:axiom} Complex numbers (Cartesian)
A complex number $z$ is written in **Cartesian form** as: 
\begin{equation*} z = a + bi \end{equation*} 
where: 
- $a$ is the real part of $z$, denoted as $\operatorname{Re}(z) = a$  
- $b$ is the imaginary part of $z$, denoted as $\operatorname{Im}(z) = b$ (where $b$ itself is a real number).

If $a = 0$, $z$ is purely imaginary whereas if $b = 0$, $z$ is purely real.
```

```{seealso} Example
Solve the quadratic equation $x^2 + 2x + 5 = 0$
```{tip} Click here for solutions
:class: dropdown
To solve this equation, we can use the quadratic formula:

\begin{equation*} x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a} \end{equation*}

For the equation $x^2 + 2x + 5 = 0$, the coefficients are:
- $a = 1$
- $b = 2$
- $c = 5$

Substituting these values into the formula:
\begin{equation*} x = \frac{-2 \pm \sqrt{2^2 - 4(1)(5)}}{2(1)}
\newline = \frac{-2 \pm \sqrt{4 - 20}}{2}
\newline = \frac{-2 \pm \sqrt{-16}}{2}\end{equation*}

Since $\sqrt{-16} = 4i$ (where $i$ is the imaginary unit, $i = \sqrt{-1}$):

\begin{equation*} x = \frac{-2 \pm 4i}{2}\end{equation*}

Then we simplify by dividing each term in the numerator by $2$ which gives

\begin{equation*} x = -1 \pm 2i\end{equation*}

Therefore, the solutions are:

\begin{equation*} x = -1 + 2i \quad \text{and} \quad x = -1 - 2i \end{equation*}

where both solutions are complex numbers, with -1 being the real parts and $\pm 2i$ are the imaginary parts. 

```


## Manipulation of Complex Numbers

### Addition, Subtraction, and Multiplication
Let $z_1 = a + bi$ and $z_2 = c + di$

**Addition:** We combine real parts and imaginary parts separately. 
\begin{equation*} z_1 + z_2 = (a + c) + (b + d)i\end{equation*}

**Subtraction:** We do much the same for subtraction:
\begin{equation*} z_1 - z_2 = (a - c) + (b - d)i\end{equation*}

**Multiplication:** This is slightly more complicated. We expand using the distributive law (i.e. how you would expand ($a + b)^2$) and substitute then $i^2 = -1$ to simplify the solution:

\begin{equation*}z_1 z_2 = (a + bi)(c + di) = ac + ad\,i + bc\,i + bd\,i^2 = (ac - bd) + (ad + bc)i \end{equation*}

```{seealso} Examples

Let $z_1 = 3 + 4i$ and $z_2 = 1 - 2i$. Work out $z_1 + z_2$,  $z_1 - z_2$ and $z_1 \times z_2$, simplifying your answers fully where required. 

```{tip} Click here for solutions
:class: dropdown
By adding $z_1$ and $z_2$ we get 
\begin{equation*}z_1 + z_2 = (3 + 1) + (4 + (-2))i = 4 + 2i \end{equation*}
Whereas for subtraction we get 
\begin{equation*} z_1 - z_2 = (3 - 1) + (4 - (-2))i = 2 + 6i \end{equation*}
For multiplication, it is best to put each complex number in brackets and expand it as follows: 
\begin{equation*}
z_1 \times z_2 = (3 + 4i)(1 - 2i) = 3 - 6i + 4i - 8i^2 = 3 - 2i + 8 = 11 - 2i
\end{equation*}
```

Division gets slightly more complicated, we need to look at **complex conjugates** to understand how to do this. 
## The Complex Conjugate
```{prf:axiom} Complex conjuage
The complex conjugate of $z = a + bi$ is denoted by $z^*$ (or $\bar{z}$):
\begin{equation*}z^* = a - bi \end{equation*} 
```
Key properties of the complex conjugate include:
- $z + z^* = 2a = 2 \times \operatorname{Re}(z)$
- $z - z^* = 2bi = 2i \times \operatorname{Im}(z)$
- $z \times z^* = (a + bi)(a - bi) = a^2 - b^2 i^2 = a^2 + b^2$ (always a non-negative real number)

```{seealso} Example
Express $(-1 - 2i) + (2 + 7i)$ in the form $x + iy$ and write down the complex conjugate. 
```{tip} Click here for solutions
:class: dropdown
First we need to simplify by adding the real and imaginary compontents separately: 
\begin{equation*}
(-1 - 2i) + (2 + 7i) = (-1 + 2) + (-2 + 7)i = 1 + 5i
\end{equation*}

Then to find the complex conjugate, we flip the sign of the imaginary part:

\begin{equation*}
(1 + 5i)^* = 1 - 5i
\end{equation*}
```
Now we've defined the complex conjugate, we can divide two complex numbers. 


## Division of Complex Numbers
To divide $z_1$ by $z_2$, you multiply both numerator and denominator by the complex conjugate of the denominator ($z_2^*$), 
\begin{equation*} \frac{z_1}{z_2} = \frac{a + bi}{c + di} = \frac{(a + bi)(c - di)}{(c + di)(c - di)} = \frac{(ac + bd) + (bc - ad)i}{c^2 + d^2} \newline
\\[1em]
\frac{z_1}{z_2} = \left(\frac{ac + bd}{c^2 + d^2}\right) + \left(\frac{bc - ad}{c^2 + d^2}\right)i 
\end{equation*}

Note that if the denominator is purely imaginary and does not a have real component, you do not need to change the sign. 

\begin{equation*} \frac{z_1}{z_2} = \frac{a + bi}{di} = \frac{(a + bi)(di)}{(di)(di)} = \frac{adi + bdi^2}{d^2i^2}
\\[1em]
\newline = \frac{adi - bd}{-d^2} = \frac{bd - adi}{d^2}
\end{equation*}

```{seealso} Example
1. Let $z_1 = 4 + 2i$ and $z_2 = 1 + 3i$. Express $z_1 \div z_2$ in the form $a + bi$. 
2. Let $z_3 = 6 + 4i$ and $z_4 = 2i$. Express $z_3 \div z_4$ in the form $a + bi$.

```{tip} Click here for solutions
:class: dropdown
1. Multiplying numerator and denominator by $z_2^* = 1 - 3i$:     
    \begin{equation*}
    \frac{z_1}{z_2} = \frac{4 + 2i}{1 + 3i} = \frac{(4 + 2i)(1 - 3i)}{(1 + 3i)(1 - 3i)} 
    \\[1em]
    \newline = \frac{(4 \times 1 + 2 \times 3) + (2 \times 1 - 4 \times 3)i}{1^2 + 3^2}
    \\[1em]
    \newline = \frac{(4 + 6) + (2 - 12)i}{1 + 9} = \frac{10 - 10i}{10} = 1 - i
    \end{equation*}
2. Recall that if there is no real component to the denominator then we simply multiply both the top and bottom of the fraction by the numerator. 
    \begin{equation*}
    \frac{z_1}{z_2} = \frac{6 + 4i}{2i} = \frac{(6 + 4i)(2i)}{(2i)(2i)} = \frac{12i + 8i^2}{4i^2}
    \end{equation*}

    \begin{equation*}
    \frac{z_1}{z_2} = \frac{12i - 8}{-4} = \frac{8 - 12i}{4} = 2 - 3i
    \end{equation*}
    
```
