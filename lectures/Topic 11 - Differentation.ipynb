# Topic 11: Introduction to Differentiation

## Chemistry motivation

When working with equations of straight lines, it is easy to find the gradient. However, when we have higher order functions (those with powers of $x$ that are greater than 1) the problem is more difficult (and interesting!) since the gradient changes depending on where you are in the function.

**Differentiation** is the mathematical procedure used to find the gradient (rate of change) of a function for any given point on the curve. 

In chemistry, we use this constantly. For example, consider the synthesis of ammonia:

\begin{equation*} \frac{1}{2}\ce{N2} + \frac{3}{2}\ce{H_2} \rightarrow \ce{NH_3} \end{equation*}

At any time $t$, the concentrations of the reactants are $[\ce{N2}]$ and $[\ce{H2}]$, and the concentration of the product is $[\ce{NH3}]$. The **rate of reaction** ($\nu_c$) is defined as the rate of change of these concentrations with respect to time.

## Gradients: Straight Lines vs. Curves

### Rate of Change of a Linear Function
For a straight line given by $y = mx + c$, the rate of change of $y$ with respect to $x$ is constant everywhere along the line. We calculate this as the **gradient**, $m$:

\begin{equation*} \text{gradient} = \frac{\text{change in } y}{\text{change in } x} = \frac{\Delta y}{\Delta x} = \frac{y_2 - y_1}{x_2 - x_1} = m \end{equation*}

Because the gradient of a straight line is the same at every point, finding its rate of change is straightforward.

### Gradient of a Curve
If a function produces a curve rather than a straight line, the gradient is constantly changing. For example, on an exponential decay curve like $h(x) = 5e^{-x}$, the steepness at $x = -1$ is completely different from the steepness at $x = 1$. 

### Notation
The resulting general expression for the gradient of a function is called the **derivative** (or **first derivative**). Depending on how a function is written, we use a few standard notations:

| Function | Derivative | Pronunciation |
| :--- | :--- | :--- |
| $y$ | $\frac{dy}{dx}$ | "d-y by d-x" or "d-y d-x" |
| $f(x)$ | $f'(x)$ | "f prime of x" or "f dash of x" |
| $f(x)$ | $\frac{d}{dx}(f(x))$ | "d by d-x of f of x" |

## Differentiating Polynomials
This section will serve as a toolkit for how to differentiate common functions that you might encounter as chemists alongside examples. 

### Simple polynomials

```{prf:axiom} Differentiating $y = x^n$
For any function of the form $y = x^n$:

 \begin{equation*}\frac{dy}{dx} = \frac{d(x^n)}{dx} = nx^{n-1} \end{equation*}

```
To summarise, to differentiate a function with respect to $x$ you **multiply by the power $n$, then reduce the power by $1$**. By convention, $x$ is often used for functions but as seen in the examples below, the rules hold regardless of the variable used. 

```{seealso} Examples
1. Differentiate $y = x^7$ with respect to $x$.
---
2. Differentiate $y = x^{-2}$ with respect to $x$.
---
3. Differentiate $y = \frac{1}{x^3}$ with respect to $x$.
---
4. Differentiate $T^{1/2}$ with respect to $T$.
---
5. Differentiate $\sqrt[3]{\alpha^4}$ with respect to $\alpha$


```{tip} Click here for solutions
:class: dropdown
1. \begin{equation*} \frac{dy}{dx} = 7x^{7-1} = 7x^6\end{equation*}
---
2. \begin{equation*} \frac{dy}{dx} = -2x^{-2-1} = -2x^{-3} = -\frac{2}{x^3} \end{equation*}
---
3. First, rewrite the fraction using a negative index ($y = x^{-3}$):
    \begin{equation*}\frac{dy}{dx} = \frac{d}{dx}(x^{-3}) = -3x^{-3-1} = -3x^{-4} = -\frac{3}{x^4} \end{equation*}
---
4.    \begin{equation*} \frac{d}{dT}(T^{1/2}) = \frac{1}{2}T^{\frac{1}{2}-1} = \frac{1}{2}T^{-\frac{1}{2}} = \frac{1}{2\sqrt{T}}  \end{equation*}
---
5. First we rewrite the root as a fractional power ($\sqrt[3]{\alpha^4} = \alpha^{4/3}$) and then differentiate: 
    \begin{equation*} \frac{d}{d\alpha}(\alpha^{4/3}) = \frac{4}{3}\alpha^{\frac{4}{3}-1} = \frac{4}{3}\alpha^{\frac{1}{3}} = \frac{4}{3}\sqrt[3]{\alpha} \end{equation*}
```

Often, functions do not only have variables and powers but they also have constants. Thankfully these are easy to deal with when differentiating using the following rules: 

* **Constants ($y = c$):** A constant does not change, meaning its graph is a flat horizontal line with zero slope.
   \begin{equation*}\frac{d(\text{constant})}{dx} = 0 \end{equation*}

* **Constant Multiples ($y = ax^n$):**  Any constant coefficient $a$ simply stays out in front and multiplies the derivative.
   \begin{equation*}\frac{d}{dx}(ax^n) = a \frac{d(x^n)}{dx} = a \times nx^{n-1} \end{equation*}
  *(For a straight line $y = mx$, this gives $\frac{d}{dx}(mx) = m$.)*

If a function consists of several terms added or subtracted together, we differentiate it **term by term**, applying the basic rules to each piece individually. Using the rules for differentiating constants and $x^n$ we can now define a rule for differentiating any polynomial: 

```{prf:axiom} Differentiating a Polynomial
Differentiate $y = ax^2 + bx + c$ with respect to $x$ (where $a$, $b$, and $c$ are constants):

\begin{equation*} \frac{dy}{dx} = \frac{d}{dx}(ax^2) + \frac{d}{dx}(bx) + \frac{d}{dx}(c) = 2ax + b + 0 = 2ax + b \\[1em] 

\end{equation*}

```

```{hint} Differentiation from first principles (non-examinable) 
:class: dropdown
Understanding differentiation from first principles allows us to see why the rules for differentiating outlined above exist. 

If we pick two points $(x_1, y_1)$ and $(x_2, y_2)$ on a curve, the ratio $\frac{\Delta y}{\Delta x}$ gives the average gradient between them, which depends heavily on both where we start and the size of $\Delta x$.

To find the exact gradient at a single point $A$ on a curve, we look at the **tangent** which is the straight line that just touches the curve at point $A$. The gradient of the curve at $A$ is equal to the gradient of this tangent.

To derive a general equation for the gradient at any point on a curve, we start with two points:
* Point $A$ with coordinates $(x, y)$
* Point $B$, which is very close to $A$, with coordinates $(x + \delta x, y + \delta y)$

*(Note: We use the lowercase Greek letter $\delta$ to represent an infinitesimally small change in a variable.)*

The straight line joining $A$ and $B$ is called a **chord**. The gradient of this chord is:

\begin{equation*} \frac{\text{change in } y}{\text{change in } x} = \frac{y_B - y_A}{x_B - x_A} = \frac{(y + \delta y) - y}{(x + \delta x) - x} = \frac{\delta y}{\delta x} \end{equation*}

As we move point $B$ closer and closer to point $A$, the distance $\delta x$ shrinks toward zero, and the gradient of the chord approaches the true gradient of the tangent at $A$. We write this rigorously using a limit:

\begin{equation*}\text{gradient} = \lim_{\delta x \to 0} \frac{\delta y}{\delta x} \end{equation*}

For example, suppose we want to derive the gradient of $y = x^2$. Let's apply this method "from first principles" to find the gradient of the curve $y = x^2$. First we define the coordinates: 
   * Point $A$ is $(x, x^2)$
   * Point $B$ is $(x + \delta x, (x + \delta x)^2)$

Then we expand the $y$-coordinate of $B$ to make simplifying easier
   \begin{equation*} (x + \delta x)^2 = x^2 + 2x\delta x + (\delta x)^2 \end{equation*}
   
and then calculate the gradient of the chord,
\begin{equation*} \frac{\text{change in } y}{\text{change in } x} = 
\newline \\[1em] \frac{(x + \delta x)^2 - x^2}{(x + \delta x) - x} = \frac{x^2 + 2x\delta x + (\delta x)^2 - x^2}{\delta x} 
\newline \\[1em] = \frac{2x\delta x + (\delta x)^2}{\delta x} = 2x + \delta x. \end{equation*}

Finally we take the limit as $\delta x \to 0$ which gives
\begin{equation*} \text{gradient} = \lim_{\delta x \to 0} (2x + \delta x) = 2x + 0 = 2x \end{equation*}

So, the general expression for the gradient of $y = x^2$ is $2x$. We can use this to find the gradient at any specific $x$-value:
* At $x = 2$: gradient $= 2(2) = 4$ (positive slope)
* At $x = -2$: gradient $= 2(-2) = -4$ (negative slope)
* At $x = 0$: gradient $= 2(0) = 0$ (flat turning point)

```

### More complex algebraic expressions
The aim is to always have your function in the form of a polynomial before you differentiate. For example before we can differentiate expressions containing brackets or fractions, we need to rearrange them into a standard sum of $x^n$ terms. 

*(Note: For complex products and fractions that cannot be simplified this way, we use the Product Rule and Quotient Rule, which we will cover later.)*
```{seealso} Example (Brackets) 
Differentiate $y = (3x - 4)(x + 5)$ with respect to $x$
```{tip} Click here for solutions
:class: dropdown
To differentiate a product of brackets, multiply out the brackets first and collect like terms.
\begin{equation*} y = 3x^2 + 15x - 4x - 20 = 3x^2 + 11x - 20 \end{equation*}
then we differentiate term by term:
\begin{equation*} \frac{dy}{dx} = \frac{d}{dx}(3x^2) + \frac{d}{dx}(11x) + \frac{d}{dx}(-20) = 6x + 11 \end{equation*}
```
```{seealso} Example (Fractions) 
Differentiate $s = \frac{t^3 - 2t^2 + 7t}{t^2}$ with respect to $t$
```{tip} Click here for solutions
:class: dropdown
To differentiate a fraction with a single term in the denominator, split it into separate fractions, cancel terms, and rewrite any remaining denominators using negative powers. For our function, splitting looks like:
 \begin{equation*} s = \frac{t^3}{t^2} - \frac{2t^2}{t^2} + \frac{7t}{t^2} = t - 2 + \frac{7}{t} = t - 2 + 7t^{-1} \end{equation*}
 \begin{equation*} \frac{ds}{dt} = \frac{d}{dt}(t) + \frac{d}{dt}(-2) + \frac{d}{dt}(7t^{-1}) = 1 - 0 + 7(-1)t^{-2} = 1 - 7t^{-2}\end{equation*}
```


```{seealso} Physical Chemistry Application: Coulomb's Law

The electrostatic interaction energy $E_{\text{elec}}(r)$ between two point charges $q_1$ and $q_2$ separated by a distance $r$ is given by:

 \begin{equation*}E_{\text{elec}}(r) = \frac{1}{4\pi\varepsilon_0\varepsilon_r}\frac{q_1 q_2}{r}\end{equation*}

where $\varepsilon_0$ is the vacuum permittivity and $\varepsilon_r$ is the relative permittivity of the medium.

In physics and chemistry, force is the negative first derivative of potential energy with respect to distance ($F = -\frac{dE(r)}{dr}$). Use differentiation to derive the expression for the electrostatic force (Coulomb's law):
```{tip} Click here for solutions
:class: dropdown
First we set up the derivative and factor out all $\textcolor{blue}{\text{constants}}$
   \begin{equation*} F = -\frac{dE_{\text{elec}}(r)}{dr} = -\frac{d}{dr}\left(\frac{1}{\textcolor{blue}{4\pi\varepsilon_0\varepsilon_r}}\frac{\textcolor{blue}{q_1 q_2}}{r}\right) = -\textcolor{blue}{\frac{q_1 q_2}{4\pi\varepsilon_0\varepsilon_r}}\frac{d}{dr}\left(\frac{1}{r}\right) \end{equation*}

Next we rewrite $\frac{1}{r}$ as $r^{-1}$ and apply the power rule giving
   \begin{equation*}\frac{d}{dr}(r^{-1}) = -1r^{-2} = -\frac{1}{r^2} \end{equation*}
and finally we substitute this back into the force equation:
   \begin{equation*}F = -\frac{q_1 q_2}{4\pi\varepsilon_0\varepsilon_r} \times \left(-\frac{1}{r^2}\right) = \frac{1}{4\pi\varepsilon_0\varepsilon_r}\frac{q_1 q_2}{r^2}. \end{equation*}
```
## Stationary points 
Differentiation allows us to calculate the gradient (slope) of a curve at any given point. A **stationary point** is a point on a function $f(x)$ where the graph temporarily stops rising or falling. This means the curve is completely flat and has a gradient of zero.

Because the first derivative, $f'(x)$ (or $\frac{dy}{dx}$), represents the gradient of the curve, we can locate all stationary points by setting the derivative equal to zero and solving for $x$:
\begin{equation*}f'(x) = 0 \end{equation*}

Once you find the $x$-values that satisfy this equation, substitute them back into the original function $f(x)$ to find their corresponding $y$-coordinates.

There are three types of stationary points: 
* local maximum (the curve goes up, flattens out at a peak, and goes down)
* local minimum (the curve goes down, flattens out at a trough, and goes up)
* point of inflection (the curve flattens out momentarily before continuing in the same direction, either continuing up or continuing down).

To determine which type of stationary point you have found, you can check the gradient just before and after the point, or use the second derivative test, $f''(x)$. The second derivative test classifies a stationary point by checking the concavity of the curve by differentiating twice. If $f''(x)$ is greater than $0$, it's a local minimum and if $f''(x)$ is less than $0$, it's a local maximum. However, if $f''(x) = 0$, the test is inconclusive, but you can just check the gradient on either side instead!

```{seealso} Example - Stationary points
Given the function:
\begin{equation*} f(x) = x^3 - 3x^2 - 9x + 5\end{equation*}
Find $f'(x)$, $f''(x)$, and hence find any stationary points and whether they are maximums or minimums. 
```{tip} Click here for solutions
:class: dropdown
First we differentiate, giving  
\begin{equation*} f'(x) = 3x^2 - 6x - 9, \end{equation*}
and then set $f'(x) = 0$,
\begin{equation*} 3x^2 - 6x - 9 = 0 \end{equation*}
then we divide through by $3$ and factorise which gets
\begin{equation*} x^2 - 2x - 3 = 0 \end{equation*}
\begin{equation*}(x - 3)(x + 1) = 0 \end{equation*}
and so we find that 
\begin{equation*} x = 3 \quad \text{or} \quad x = -1. \end{equation*}

Next we substitute both $x$-values back into the original function $f(x)$ to find their $y$-coordinates:
At $x = -1$: $f(-1) = (-1)^3 - 3(-1)^2 - 9(-1) + 5 = -1 - 3 + 9 + 5 = 10 \implies \mathbf{(-1, 10)}$ and at $x = 3$: $f(3) = (3)^3 - 3(3)^2 - 9(3) + 5 = 27 - 27 - 27 + 5 = -22 \implies \mathbf{(3, -22)}.$

Finally we classify both points using the second derivative test by differentiating a second time:
\begin{equation*} f''(x) = 6x - 6 \end{equation*}

and evaluate $f''(x)$ at each stationary point, so at $x = -1$, $f''(-1) = 6(-1) - 6 = -12$ (less than $0$) $\implies$ $(-1, 10)$ is a local maximum. Whereas at $x = 3$: $f''(3) = 6(3) - 6 = 12$ (greater than $0$) $\implies$ $(3, -22)$ is a local minimum.
```
