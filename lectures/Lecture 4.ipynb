

#  Topic 4: Further error calculations

## Chemistry motivation
It is not only important to calculate expermimental errors, but to be able to perform additional calcuations with them, as well as be able to interpret the results of your calulations. For example, we may need to calculate a quantity that cannot be measured directly but is a known function of two or more quantities that we can measure and it is important that we are able to quantify our uncertainty about that quantity.

```{seealso} Example
In the physical chemistry laboratory the standard Gibbs free energy change $\Delta G$ is related to the equilibrium constant $K$ and absolute temperature $K$ using the following equation:
\begin{equation*}\Delta G = - RT\ln(K),\end{equation*}

such that R is the ideal gas constant, which is equal to 8.3114 J mol$^{-1}$ Kelvin$^{-1}$. Here, $\Delta G$ is a function of two independent variables, $T$ and $K$, each with an associated uncertainty arising from random errors in measurement. 

Suppose 
\begin{equation*}T = 300 \pm 2 \text{ Kelvin}\end{equation*} and \begin{equation*} K = 1995 \pm 49\end{equation*}
We can calculate $\Delta G$ easily as $- 8.3145 \times 300 \times \ln(1995) = -18950$. But now the question comes, how do the uncertainties in our independent variables $T$ and $K$ influence the uncertainty in of the dependent variable $\Delta G$? 

We will return to answer this, but first we need some theory!

This rule extends to the addition of more than two quantities, and also to subtraction. For
example, if $Z = A + B - C$,
Then
$\sigma_z = \sqrt{\sigma_a^2 + \sigma_b^2 + \sigma_c^2}$

```{hint} But why? (Non-examinable)
:class: dropdown
Recall that error is normally reported using the standard deviation, $\sigma$, and that variance is simply $\sigma^2$.   

Let $Z = nX + mY$ where n and m are constants with no associated error. Let $Var(Z)$ represent $\sigma_z^2$. In probability theory, it is known that

\begin{equation*} \sigma_{mX+nY}^2 = Var(mX+ nY) = m^2Var(X) + n^2Var(Y) \implies \end{equation*}
\begin{equation*}\sigma_{mX+nY} = \sqrt{ m^2\sigma_x^2 + n^2\sigma_y^2} \end{equation*}


Now, let us suppose that $m=n=1$, giving $Var(X + Y)$. Using the formula above we have 
\begin{equation*}
Var(X+Y) = (1)^2Var(X) + (1)^2Var(Y) = \end{equation*}
\begin{equation*}1^2\sigma_x^2 + 1^2\sigma_y^2 = \sigma_x^2 +  \sigma_y^2 \end{equation*} 
which tells us that $\sigma_{x+y} = \sqrt{\sigma_x^2 = \sigma_y^2}$
whereas 
\begin{equation*}Var(X - Y) = (1)^2Var(X) + (-1)^2Var(Y) = \end{equation*}
\begin{equation*}1^2\sigma_x^2 + (-1)^2\sigma_y^2 = \sigma_x^2 + \sigma_y^2,\end{equation*}
which is why you get the same formula for propagatig errors regardless of whether you are adding or subtracting the experimental values.

```{warning} Note
This only works if your two experimental measurements are independent of eachother!

```


```{seealso} Example
In a titration experiment the volume dispensed from a burette is measured as the difference between the initial ($x$) and final ($y$) reading on the burette. If the scale on the burette consists of 1 mL graduations, one might estimate the uncertainty of each
initial and final reading as $\pm 0.5$ mL.

So we have an initial Volume = $25 \pm 0.5 \text{cm}^3$ and final Volume = $5 \pm 0.5$ cm$^3$
Estimate the standard deviation in the volume dispensed, $\sigma_V$.
```{note} Click here to see solutions
:class: dropdown
From the question, we have $V = x - y$, $\sigma_x = 0.5$ and $\sigma_y = 0.5.$

Applying our general addition/subtraction rule for the propagation of errors,
\begin{equation*}\sigma_𝑉 = \sqrt{0.5^2 + 0.5^2} = 0.71\text{cm}^3, \end{equation*}
and
\begin{equation*} \text{Dispensed Volume} = 20 \pm 0.7 \end{equation*}
```



### Multiplication and divison
Let 𝑍 represent a quantity that we wish to calculate by multiplication (or division) of
two experimental measurements, $A$ and $B$, with random error $\sigma_a$ and $\sigma_b$, respectively.
$Z = A \times B$, $(Z \pm z) = (A \pm a) \times (B \pm b)$
To evaluate the error, $\sigma_z$, associated with this new quantity, we now work with the
relative error.

```{prf:axiom}  Propagation of Uncertainties for Random, Independent Errors: Multiplication
The relative error in $Z$ is

\begin{equation*} \frac{\sigma_z}{Z} = \sqrt{\left(\frac{\sigma_a}{A}\right)^2 + \left(\frac{\sigma_b}{B}\right)^2} \end{equation*}

```
As previous, this rule extends to the multiplication or division of more than two
quantities. For example, if $Z = \frac{A \times B}{C}$, then
\begin{equation*} \frac{\sigma_z}{Z} = \sqrt{ \left(\frac{\sigma_a}{A}\right)^2 + \left(\frac{\sigma_b}{B}\right)^2 + \left(\frac{\sigma_c}{C}\right)^2} \end{equation*}


```{seealso} Example

In a sheet metal factory producing rectangular sheets of aluminium, the mean length, $\bar{x}$ of multiple samples is measured as 20 m with a standard deviation 0.2 m , and the
mean width, $\bar{y}$, of 10 m with a standard deviation 0.15 m.
Estimate the mean area and standard deviation, $\sigma_Z$.

```{tip} Click here to see the solution
:class: dropdown
The area is $Z = x \times y$, so the mean area is`
\begin{equation*} \bar{Z} = \bar{x} \times \bar{y} = 20 × 10 = 200 \text{m}^2 \end{equation*}

Applying our general rule for the propagation of errors involving multiplication our relative error is
\begin{equation*} \frac{\sigma_z}{Z} = \sqrt{\left(\frac{0.2}{20}\right)^2 + \left(\frac{0.15}{10}\right)^2} \end{equation*}
Therefore our standard deviation is $\sigma_Z = 200 \times 3.25 \times 1 = 3.6 \text{m}^2$ and we can report our area as $Z = (200 \pm 3.6)\text{m}^2$.

```
