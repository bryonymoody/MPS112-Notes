# Topic 7: Exponentials and logarithms

## Chemistry motivation
You will often hear of things increasing at an "exponential rate" to imply an incredibly fast rate of increase. You observed this is chemistry in a variety of scenarios, such as the increase in the rate of a chemical reaction when the temperature is increased, or the increasing number of neutrons formed during a nuclear chain reaction. Therefore it is important that you can work comfortably with equations that involve exponentials. 

## Mathematical theory

### Terminology recap
Equations typically involve:
- constants (i.e. quantities with a fixed and pre-determined value)
- and variables (i.e. quantities that may take on any one of a range of values).

Constants and variables can be:
 - Latin letters, such as $x$, $y$, $n$, $m$, $T$, $H$, $S$
 - Greek letters, such as $\alpha$, $\beta$, $\lambda$, $\phi$, $\theta$.
 
For example, let us revisit the relationship between the Gibbs energy of a reaction and the equilibrium constant
\begin{equation*}
\Delta G = –RT \ln K
\end{equation*}
such that we have the following variables:
- $\Delta G$: Gibbs energy 
 - $T$: Temperature (in Kelvin) 
 - $K$: equilibrium constant, 
 
whereas $R$ is the universal gas constant is a constant = $8.31451 \text{Jmol}^{-1}\text{K}^{-1}$. Finally, we have $\ln$ which is the symbol for natural log, we will cover what this is in this topic. If we measure $K$ as a function of temperature, we can use the above equation to calculate $\Delta G$.

### Rearranging equations
Often when working we equations, it is helpful to be able to rearrange it to make a different variable the subject of the equation. 

```{seealso} Example
Given the ideal gas equation $pV = nRT$, we can rearrange this in various ways, depending on which variable we can interested in calculating. For example, by divding the right had side my $V$ we make $p$ the subject of the equation: 

\begin{equation*}
p = \frac{nRT}{V}
\end{equation*}
Similarly, we can obtain the following versions of the ideal gas equation
\begin{equation*}
V = {nRT}, \\[1em]
T = \frac{pV}{nR},
\newline \\[1em]
\text{and}
\newline \\[1em]
n = \frac{pV}{RT}.

\end{equation*}
```

Equations of the form 
\begin{equation*}
\textcolor{blue}{y} = \textcolor{green}{a}^{\textcolor{red}{x}}
\end{equation*}

such $\textcolor{blue}{y}$ is the subject of the equation, $\textcolor{red}{x}$ is the exponent and $\textcolor{green}{a}$ is a positive constant refered to as the base of the equation.

Exponentional equations differ from equations of the form
\begin{equation*}
\textcolor{blue}{y} = \textcolor{red}{x}^{\textcolor{green}{a}}
\end{equation*}
that you might have encoutered before. In the equation above $x$ varies while the exponent $a$ is constant. Whereas, in exponential equations, the exponent is now $x$ and so that varies, while the base $a$ stays constant. 

When rearranging exponential equations things are a little bit trickier since $x$ is the exponent, but this is where logarithms come in, denoted as $\log_a$. Logarithms are the inverse of exponentials, much like multiplicaition is the inverse of division. 

If we want to rearrange 
\begin{equation*}
\textcolor{blue}{y} = \textcolor{green}{a}^{\textcolor{red}{x}}
\end{equation*}
to make $x$ the subject then we take the log of both sides giving 

\begin{equation*}
\log_{\textcolor{green}{a}}(\textcolor{blue}{y}) = \log_{\textcolor{green}{a}}(\textcolor{green}{a}^{\textcolor{red}{x}}) \\[1em]
\log_{\textcolor{green}{a}}(\textcolor{blue}{y}) = \textcolor{red}{x}.
\end{equation*}

Therefore, as long as the base of the logarithm matches the base of the equation, then the log function and $\textcolor{green}{\text{base}}$ "disappear" and only the $\textcolor{red}{\text{exponent}}$ is left behind. 

```{seealso} Examples
Given 
\begin{equation*}
\textcolor{blue}{y} = \textcolor{green}{a}^{\textcolor{red}{x}}
\end{equation*}

we see that

\begin{equation*}
\textcolor{blue}{8} = \textcolor{green}{2}^{\textcolor{red}{3}}
 \text{ so } \log_{\textcolor{green}{2}}(\textcolor{blue}{8} = \textcolor{red}{3} ) \end{equation*}

Now suppose we have

\begin{equation*}
\textcolor{blue}{3.16} = \textcolor{green}{10}^{\textcolor{red}{x}}
\text{ so } \textcolor{red}{x} = \log_{\textcolor{green}{10}}(\textcolor{blue}{3.16}).
\end{equation*}
```

### The natural logarithm
There is a special case of an exponential function where the base $a$ is equal to $e$ (which is known as Eulers number, approximately equation to 2.718). The inverse function for equations of the form
\begin{equation*}
\textcolor{blue}{y} = \textcolor{green}{e}^{\textcolor{red}{x}} 
\end{equation*}

would be $\log_{\textcolor{green}{e}}(\textcolor{blue}{y})$ which we denote as $\ln(\textcolor{blue}{y})$. However all rules for working with $\log_a$ hold for working with $\ln$.

### Rules for working with logarithms
Now it will be helpful to define what these rules are. Much like working with indicies/exponents (as seen in section X), there are a set of rules that help us with working with logarithms. 

```{prf:axiom} Logarithms
The rules for working with logarithms are as follows:
- $\log_a(x \times y) = \log_a(x) + \log_a(y) \newline$ 
  $\textcolor{blue}{\text{Example: } \log_4(7 \times 5) = \log_4(7) + \log_4(5)} 
  \\[1em] $
- $\log_a(\frac{x}{y}) = \log_a(x) - \log_a(y) \newline$
  $\textcolor{blue}{\text{Example: } \log_{10}(\frac{10}{2}) = \log_{10}(10) - \log_{10}(2)} 
  \\[1em]$
- $\log_a(x^b) = b \times \log_a(x) \newline$
  $\textcolor{blue}{\text{Example: } \log_{10}(5^3) = 3\log_{10}(5)} 
  \\[1em] $
- $\log_a(b) = \frac{\log_c(b)}{\log_c(a)} \newline$
  $\textcolor{blue}{\text{Example: } \log_2(3) = \frac{\log_{10}(3)}{ \log_{10}(2)}} 
  \\[1em]$
```

### Conversion between $\ln$ and $\log_{10}$
Often, many equations in chemistry are of the form $\ln$, but it can be helpful to covert to $\log_{10}$ since humans find thinking in power of 10 more intuitive. 

```{prf:axiom} Conversion from base $e$ and $10$
We can covert using the following equations
\begin{equation*} \ln(x) \approx 2.303 \log_{10}(x) \end{equation*}
or similarly
\begin{equation*} \log_{10}(x) \approx 0.434 \times \ln(x) \end{equation*}

```{tip} But why? (Non examinable)
:class: dropdown
Recall that  $\log_a(b) = \frac{\log_c(b)}{\log_c(a)}$. We can use this to say that

\begin{equation*}
\log_{10}(x) = \frac{\ln(x)}{\ln(10)} \implies \\[1em]
\ln(10) \times \log_{10}(x) = \ln(x) \\
\end{equation*}
and since $\ln(10) = 2.302585...$
we can say that 
\begin{equation*}
2.303 \times \log_{10}(x) \approx \ln(x) 
\end{equation*}
```
