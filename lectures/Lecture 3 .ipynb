# Topic 3 - Calculating experimental errors

## Chemistry motivation
Errors are an inherent part of any chemical analysis and impact subsequent calculations. This does not mean that a measurement is necessarily incorrect but reflects that it is rarely possible to know the true value of any measurement or quantity determined experimentally. When making quantitative measurements within the laboratory, you must take into consideration that the measurement will generally differ from the ‘true’ value due to inevitable experimental errors.  

Three types of error may be defined:
- **Gross personal error**:	Mistake by the practitioner in when performing the experiment, or by instrument malfunction. Such errors may be possible to detect by repeated measurements or by comparison with independent data sets.
-  **Systematic error**: Are those that alter a result from the ‘true’ value in a reproducible way.  Such errors may arise from personal error and prejudice, but more commonly arise due to incorrect calibration of an instrument.
- **Random error**: These errors affect results in a non-reproducible way and arise from random fluctuations in the measurement process.  For example, measurements of volume dispensed by a burette may depend on the position of the eye relative to the scale (the parallax effect), while values of the concentration of a solution prepared for a titration experiment may fluctuate randomly for each practitioner.  Such errors may be reduced with increasing skill and experience.

![](erros_vis.png)
Here we focus on the case of random errors and assume that the two other sources of error have been eliminated by careful preparation and execution of the experiment.

```{note} An interesting side note
**Systematic error example – 1970s Himalayas**

Original altitude measurements of Himalayan peaks were consistently too high, due to systematic mis-calibration of instruments. The surveyers failed to consider the mass of the mountains themselves when determining the acceleration due to gravity, a quantity interpreted by instruments to calculate altitude. The errors were only discovered when radar-surveying methods using orbiting satellites became common.

```

## Mathematical theory

All experimental measurements are subject to errors. We may only minimise these, and it is important to be able to quantify such errors, especially since errors can propagate!

### Presentation of errors
There are two common approaches for estimating the error in a measurement. First you might use **personal judgement**, where you estimate the precision of the measurement based on the scale used or the specification of the instrument. For example
\begin{equation*} \text{Case A: Volume}= 6.0 \pm 0.5 \text{cm}^3 \end{equation*}
\begin{equation*} \text{Case B: Volume}= 6.2 \pm 0.2 \text{cm}^3 \end{equation*}


![](error_mes.png)

Both of these measurements agree within experimental error and are acceptable to report. 

The other common approach is **repeated measures**. This can be used to estimate the experimental error. If we have repeated measurements of a quantity, this reveals a distribution of values. 

A **probability distribution** assigns a probability to every possible value a random variable can take. There are many types of prrobability distributions that are appropriate for modelling specific scenarios. Suppose we make repeated measurements of the equilibrium constant, $K$, for a chemical reaction.  Due to random errors, each value obtained may deviate slightly from the ‘true’ value.  If we plot the results as a histogram, we gain an impression of the distribution of measurements due to such errors.  
Hypothetically, if we are to take an **infinite number** of measurements and plot the results using infinitesimally narrow intervals, we obtain a **continuous distribution** for the values of K (dashed line)

![](parent_dist.png)

This hypothetical distribution is called the **parent distribution** (of $x$). If this distribution is known for a particular experiment, one could then calculate the exact probability that a measurement will lie within a specified range of $x$. 

### The Normal distribution

A remarkable result of error analysis is that the parent distribution of measurements subject to random errors has the **same characteristic shape**, regardless of the details of the experiment. The shape is known as the normal distribution, which has this characteristic "bell" shape. It depends on two parameters, $\mu$ and $\sigma$, which are estimated from the data. 

```{prf:axiom} Probability distribution function of the Normal distribution
To calculate the probability of observing a specific value $x$, you use the probability distribution function for the Normal distribution which is 
\begin{equation*}
\\[1em]
f(x) = \frac{1}{\sqrt{2\pi\sigma ^2}} \exp \left[- \frac{(x-\mu)^2}{2\sigma^2} \right]
\end{equation*}
```
In this module it is not important to memorise the equation describing the normal distribution.  Rather, we should understand qualitatively what these parameters represent:

- $\mu$ is the **population mean** and is commonly regarded as the 'true' value of  and is the point about which the distribution is **symmetrical**.
- $\sigma$ is the **standard deviation**, which measures how spread out the data is from the mean. The smaller the value of $\sigma$, the narrower the range of measurements. 

Below is a plot of a normal distribution with $\mu = 2000$ and $\sigma = 100$ 

![](norm_dist.png)

For our previous example involving measurements of the equilibrium constant, $\mu = 2000$ , $\sigma = 100$, it is common to report the uncertainty as $1\sigma$:
\begin{equation*}K = 2000 \pm 100 \end{equation*}

### Confidence intervals

The normal distribution has the useful property that, for the population, $68\%$ of the data lies within one standard deviation of the population mean. Similarly, $95\%$ of the data lies within two standard deviation of the population mean. To summarise our confidence in measurements, we provide confidence intervals. For example, a $95\%$ confidence interval would be given as $(x - 2\sigma, x + 2\sigma)$.  

### Population vs sample

Where an infinite number of measurements are made the distribution describes directly the **population**. In practical terms it is only possible to take a finite number of measurements, a **sample**. The resulting distribution of values approximates to the normal distribution.

In experiments we take a limited number of measurements. Making the assumption that the sample data are normally distributed (random errors) we can estimate $\mu$ and $\sigma$ from a finite number of measurements $N$.

The **sample mean**, $\bar{x}$, of a set of $N$ measurements is 
\begin{equation*} \\[1em]
\bar{x} = \frac{x_1 + x_2 + x_3 ... + x_N}{N} = \frac{1}{N}\sum^n_1 x_i  \end{equation*}

The **sample standard deviation**, $s$, is defined as 
\begin{equation*}s  = \sqrt{\frac{1}{N-1}\sum^N_1 (x_i - \bar{x})^2}. \end{equation*}

The **sample variance** is simply $s^2$.

To determine the equilibrium constant, K, of a reaction, a student performs 4 measurements:
