# Topic 8: Trigonometry

## Chemistry motivation
Chemistry is a 3 dimensional subject; molecules have shape which is
defined by bond angles and bond lengths. Similarly crystal structures
can be thought of as a regular 3 dimensional pattern of spheres inside
a 3 dimensional geometric shape.

![](molecule.png)
Hence, the need for geometry and trigonometry.

## Mathematical theory

In geometry, when working with triangles you are often posed with two types of problem. First, you might know the length of two sides and wish to find the length of the remaining side. In this case you need to use **pythagoras' theorem**, whereas if you have two known sides and wish to find the size of an angle, or a known side and angle and wish to find the value of an unknown side, then both these scenarios require **trigonometry.**

```{prf:axiom} Pythagoras' theorem 
Given a right angled triangle with sides $a$, $b$ and $c$ such that $c$ is the hypotenuse (longest side) then: 
\begin{equation*}
a^2 + b^2 = c^2
\end{equation*}

```{seealso} Example
Suppose we have the following triangle
![](triangle.png)
Use Pythagoras' theorem to find the missing side
```{tip} Click here for solution
:class: dropdown
To find $a$ we rearrange 
\begin{equation*}
a^2 + b^2 = c^2
\implies a^2 = c^2 - b^2 
\implies a = \sqrt{c^2 - b^2}
\end{equation*}
Substituting the side lengths in gives

\begin{equation*} a = \sqrt{5^2 - 4^2} = \sqrt{25-16} = \sqrt{9} = 3\end{equation*}



```

### Trigonometry

If at two sides of a triangle are determined, or similary if one side and one angle are determined, then we can work out all sides and angles in the triangle. First we will look at the easier case of right angled triangles. 

Right angled triangles are labelled as follows:
![](trigtri.png)

We can work out any given angle using **trigonometric ratios**
```{prf:axiom} trigonometric ratios
Given an angle $\theta$ such that $ 0 < \theta \leq 90 \deg$ we have the following three trigonometric ratios:
\begin{equation*}\sin (\theta) = \frac{\text{opposite}}{\text{hypotenuse}} \end{equation*}
\begin{equation*} \cos(\theta) = \frac{\text{adjacent}}{\text{hypotenuse}} \end{equation*}
\begin{equation*}\tan (\theta) = \frac{\text{opposite}}{\text{adjacent}} \end{equation*}
```
There are also useful trigonometric identities we often utilise when working with triangles, the two most commonly used are 
\begin{equation*} \tan\theta = \frac{\sin\theta}{\cos\theta} \end{equation*}
and
\begin{equation*} \sin^2\theta + cos^2 \theta = 1 \end{equation*}

### Sin and cosine rules

Pythagoras’ theorem states $a^2 + b^2 = c^2$ for a right-angled triangle. But what about triangles that do not contain a 90° angle?

```{prf:axiom} Cosine rule
Consider the triangle $ABC$:
![](nonright_tri.png)

The three sides are of length $a$, $b$, and $c$. The cosine rule gives us a relationship between $a$, $b$, $c$

\begin{equation*} c^2 = a^2 + b^2 - 2ab \cos C \end{equation*}

```{hint} But why? (Non examinable)
:class: dropdown

![](cosine_rule.png)

First, using sine and cosine for triangle ACD we know that 
\begin{equation*}h = b \sin C \end{equation*}
\begin{equation*}k = b \cos C \end{equation*}

The line $AD$ (height) intersects side $a$ at right angles at point $D$. Line $AD$ divides the triangle $ABC$ into two right-angled triangles, $ABD$ and $ACD$. Denote the length of $AD$ as h, $CD$ as $k$ and $BD$ as $(a – k)$.

Substituting for $h$ and $k$ from triangle ACD we can use Pathagoras' theorem to find that
\begin{equation*} \\[1em]c^2 = (b \sin C)^2 + (a - b \cos C)^2 \end{equation*}

and similarly for triangle ABD we know:
$c^2 = h^2 + (a - k)^2$

Substituting for $h$ and $k$ from triangle $ACD$ gives
\begin{equation*} \\[1em] c^2 = (b \sin C)^2 + (a - b \cos C)^2 \end{equation*}

Expanding the terms
\begin{equation*} \\[1em] c^2 = b^2 \sin^2 C + a^2 - 2ab \cos C + b^2 \cos^2 C \end{equation*}
\begin{equation*} \\[1em]c^2 = b^2 (\sin^2 C + \cos^2 C) + a^2 - 2ab \cos C \end{equation*}

Finally, recall that $\sin^2 \theta + \cos^2 \theta = 1$, therefore
\begin{equation*}\\[1em] c^2 = b^2 (1) + a^2 - 2ab \cos C \end{equation*}
which simplifies to give the cosine rule
\begin{equation*}\\[1em] c^2 = a^2 + b^2 - 2ab \cos C. \end{equation*}

For the other sides in the same triangle, the cosine rule can also be written as
\begin{equation*} \\[1em] a^2 = b^2 + c^2 - 2bc \cos A \end{equation*}
\begin{equation*} \\[1em] b^2 = a^2 + c^2 - 2ac \cos B \end{equation*}

Notice that if the angle is $\frac{\pi}{2} (90°)$, $\cos \frac{\pi}{2} $ is zero and the cosine rule simplifies to
$c^2 = a^2 + b^2$ (Pythagoras’ theorem). 

```
```{seealso} Example
Using the cosine rule

In $\triangle ABC$, $BC = 3.5$cm, $AC = 4.5$cm and $C=61°$. Find the length of $AB$.

```{tip} Click here for solutions
:class: dropdown

First, it is helpful to draw and label your triangle. By convention side $a$ is opposite angle $A$, side $b$ is opposite angle $B$ and side $c$ is opposite angle $C$. 

![](cosine_ex.png)
We are trying to find the length of $c$, therefore we use the following cosine rule:
\begin{equation*}\\[1em] c^2 = a^2 + b^2 - 2ab \cos C. \end{equation*}

Substituting our values in gives
\begin{equation*}\\[1em] c^2 = 2.5^2 + 4.5^2 - 2(3.5)(4.5) \cos 61 \end{equation*}
\begin{equation*}\\[1em] c^2 = 2.5^2 + 4.5^2 - 2(3.5)(4.5)(0.4848) \end{equation*}
\begin{equation*}\\[1em] c^2 = 17.23 \implies c = 4.15\end{equation*}



```
```{prf:axiom} Sine rule
Consider the triangle $ABC$:
![](nonright_tri.png)

The three sides are of length $a$, $b$, and $c$. The sine rule gives us a relationship between $a$, $b$, $c$ and their corresponding angles $A$, $B$ and $C$ as

\begin{equation*} \frac{a}{\sin A} = \frac{b}{\sin B} = \frac{c}{\sin C}. \end{equation*}

```{hint} But why? (Non examinable)
:class: dropdown
Again, we consider the triangle below
![](cosine_rule.png)
For the right angled triangle $ADC$ we have
\begin{equation*} \sin C = \frac{h}{b} \end{equation*}
therefore,
\begin{equation*} h = b \sin C \end{equation*}

For the right angled triangle $ADB$ we have
\begin{equation*} \sin B = \frac{h}{c} \end{equation*}
therefore,
\begin{equation*} h = c \sin B. \end{equation*}
We can combine the equations, giving 
\begin{equation*} b\sin C = c \sin B \end{equation*}
and so rearranging gives

\begin{equation*} \frac{c}{\sin C} = \frac{b}{\sin B} \end{equation*}

and you can use this same method to show that

\begin{equation*} \frac{c}{\sin C} = \frac{a}{\sin A} \end{equation*}
and as such, we obtain the sine rule. 
```

```{seealso} Example
Suppose we have a triangle $\triangle ABC$, $BC = 2.5$cm, $A = 43°$ and $B=61°$. Find the length of $AC$.

```{tip} Click here for the solution
:class: dropdown
Angles $A$ and $B$, and side $a$ are known, therefore the two fractions we select from the sine rule are

\begin{equation*} \frac{a}{\sinA} = \frac{b}{\sin B} \newline
\implies \frac{2.5}{\sin 43} = \frac{b}{\sin 61} \newline
\implies b = \frac{2.5 \times \sin 61}{\sin 43} = \frac{2.5 \times 0.8746}{0.6819} = 3.2 \end{equation*}

Therefore $AC = 3.2 $cm
```
The appropriate rule to use depends on both the information you have, and the information you would like to know. <br>
**Scenario 1:** You have two angles and a side, you want to know an angle $\implies$ sine rule. <br>
**Scenario 2:** You have two sides and angle, you want to find an angle $\implies$ sine rule, <br>
whereas <br>
**Scenario 3:** You have two sides and an angle, you want to find a side $\implies$ cosine rule.  

```{seealso} Example
In the amine $\text{RNH}_2$ molecule shown in this slide, the N – H bond length is 103.8 pm (picometres) and the H – N – H bond angle 109.3°. Calculate the separation between the hydrogen atoms.

```{tip} Click here for solutions
:class: dropdown
We need to use the cosine rule,
\begin{equation*} a^2 = b^2 + c^2 - 2bc \cos A \end{equation*}

![](nh_mol.png)

First we start by simplifying the expression, since the H – N – H bonds form an isosceles triangle, so $b = c$ and we can rewrite the cosine equation as
\begin{equation}\\[1em] a^2 = b^2 + b^2 - 2bb \cos A \newline \\[0.3em]  
a^2 = 2b^2 - 2b^2 \cos A \newline  \\[0.3em]  
a^2 = 2b^2(1 - \cos A) \newline  \\[0.3em]  
a = b\sqrt{2(1 - \cos A)}. \end{equation}

Substituting in our values gives
\begin{equation*} 
a = 103.9 \sqrt{2(1 - \cos 109.3^\circ)} \newline  \\[0.3em]  
= 103.9 \sqrt{2(1 - (-0.3305))} \newline  \\[0.3em]  
= 103.9 \sqrt{2 \times 1.3305} \newline  \\[0.3em]  
= 103.9 \sqrt{2.661} \newline  \\[0.3em]  
a = 103.9 \times 1.63 = 169.5\text{ pm} \end{equation*}
