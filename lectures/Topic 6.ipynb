## Mathematical theory

In this course we will used the method of **linear least squares** to fit a function to experimental data. In this section, we will cover how to fit a straight line to your experimental data in the optimal way. Recall that the equation of a straight line is of the form $y=mx+c$. 

```{prf:axiom} Finding the optimal gradient and intercept
To find the equation of a straight line that best fits the data, we find our gradient and intercept using the following formulae:
\begin{equation*}
m = \frac{cov(x,y)}{\sigma_x^2} 
\end{equation*}
and 
\begin{equation*}
c = \bar{y} - m\bar{x}.
\end{equation*}
**Note:** For samples, you would calculate the sample covariance and sample variance to calculate $m$. 
```
```{tip} But why? Non examinable
:class: dropdown
Optimal fitting of a straight line to the data requires the minimisation of the sum of the squared deviations (or residuals), such that we define a squared deviation to be
\begin{equation*}
d^2 = (y_i - (mx_i + c))^2.
\end{equation*}
This can be interpreted as the squared difference between the observed $y$ values and the predicted $y$ values that result from the linear equations, and the goal is to choose values for $m$ and $c$ reduce that difference as much as we possibly can. 

We will come back to the method for minimising these residuals, but we need to cover **differentiation** first.
```
```{seealso} Example
Let us return to the example in the previous topic where we examined the hours students spent revising and their subsequent test scores.

We are looking for an equation of the form $y = mx + c$ which best represents the relationship
\begin{equation*}
\text{test score} = m(\text{revision hours}) + c,
\end{equation*}
with respect to our observed data. 

We found previously that the covariance for this dataset was 63.6 and the sample standard deviation $s_x = 2.61$. So we find that
\begin{equation*}
m = \frac{cov(x,y)}{s_x^2}  = \frac{63.6}{2.61^2} = 9.35
\end{equation*}
Now we have found $m$ we can find $c$ using 
\begin{equation*}
c = \bar{y} - m\bar{x} = 53.5 - 9.35 \times 4 = 16.1 
\end{equation*}
Therefore our equation that minimises the squared residuals is 
\begin{equation*}
y = 9.35 x + 16.1
\end{equation*}
