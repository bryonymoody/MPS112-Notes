## Topic 13 - Introduction to integration


## Mathematical theory
In the previous two topics, we covered differentiation, which is the process of finding the gradient (rate of change) for any given function. Integration is the opposite process to differentiation, much like mulitplication is the opposite to division. 

If you integrate a gradient function ($\frac{dy}{dx}$) you will obtain the function of the curve, $f(x).$ If you integrate the function $f(x)$ itself, then you obtain a function that allows you to calculate the area under the curve at any given point. 

### Notation
Much like with differentiation, we need to introduce some new notation before we can begin to integrate functions: 

| Notation | Pronunciation |
| :--- | :--- |
| $\int$ | "integral" |
| $dx$ |  "d-x" |
| $\int^a_b$ | "integral between $x = b$ and $x = a$" |
| $\int f(x) \text{ } dx$ | "integral of the function $f(x)$ with respect to $x$" |


### Integration of functions

To differentiate a power of $x$, we multiply by that power, then reduce the power by 1. As integration reverses the operation of differentiation, to integrate a power of $x$, we must divide by that power $+ 1$, then increase that power by 1. We must also add a constant $c$ to the resulting function 


```{prf:axiom} Integrating $y = x^n$
For any function of the form $y = x^n$ such that $n \neq -1$:

\begin{equation*} \int x ^ n \,dx = \frac{1}{n+1}x^{n+1} + c \end{equation*}

```
To consider why we need the constant $c$, consider the following examples of differentiation:

\begin{equation*}
\frac{d}{dx}(x^2) = 2x
\end{equation*}

\begin{equation*}
\frac{d}{dx}(x^2 + 2) = 2x
\end{equation*}

\begin{equation*}
\frac{d}{dx}(x^2 + 4) = 2x
\end{equation*}


Irrespective of the value of the constant $c$, when we differentiate $x^2 + c$ with respect to $x$, the derivative is always $2x$ because the gradients of these curves are the same (gradient $= 2x$).

Therefore, the result of integrating $2x$, which is called the integral of $2x$, is not a unique function but is of the form:

\begin{equation*}
\int 2x \,dx = x^2 + c
\end{equation*}

where $c$ is called the integration constant. This is the case for the integration of any function: there is always an integration constant, and it must always be included when integrating a function. You can find the value of the constant by substituting any given point on the curve into the function, and we will look at some examples of this later on. 


#### The Special Case $x^{-1}$

Using the general rule for integrating $x^n$ for $x^{-1}$ gives an undefined result since

\begin{equation*}
\\[1em]
\int x^{-1} \, dx = \frac{x^{-1+1}}{-1+1} + c = \frac{x^0}{0} + c = \infty
\end{equation*}

To integrate $x^{-1}$, recall the differentiation of a natural logarithm:

\begin{equation*}
\frac{d}{dx}(\ln x) = \frac{1}{x}
\end{equation*}

Therefore, the rule for integrating $x^{-1}$ is:

\begin{equation*}
\int x^{-1} \, dx = \int \frac{1}{x} \, dx = \ln |x| + c
\end{equation*}

Now that we know how to to integrate any power of $x$, let us look at some examples.
```{seealso} Example
Integrate with respect to $x$ the function $x^5$.
```{tip} Click here to see solutions
:class: dropdown
First we increase the power of $x$ by 1, $x^{5+1} = x^6$, then we divide by the new power, $\frac{1}{6}x^6$. Finally we add the integration constant $+ c$ giving 
\begin{equation*}
\int x^5 dx = \frac{1}{6}x^6 + c
\end{equation*}
```

```{seealso} Example

```
### Integrating constants
Much like with differentiation, we need to know how to approach constants:
* **Constants : y = a** To integrate constants we use the property $x^0 = 1$ (where $x$ is any variable or constant) to rewrite the constant as $a = a \times 1 = a \times (\text{variable})^0.$ For example, 
    \begin{equation*}
\int a \, dx = \int a  x^0 \, dx = ax +  c.
\end{equation*}

* **Constant Multiples ($y= bx^n$):** To deal with multiplication by a constant ($y = bx^n$) we take the constant outside of the integral and deal with the constant as we usually would: 

\begin{equation*}
\int bx^n  \, dx = b \int x^n \, dx = b \frac{x^{n+1}}{n+1} + c
\end{equation*}

```{prf:axiom} Integrating polynomials (such as $dx^2 + ex + f$) 
We integrate term by term, applying the basic rule to each term:

\begin{equation*}
\int (dx^2 + ex + f) \, dx = \int dx^2 \, dx + \int ex \, dx + \int f \, dx \newline \\[1em]= \frac{d}{2 + 1}x^{2+1} + \frac{e}{1+1}x^{1+1} + \frac{f}{1+0} + c \newline \\[1em]
= \frac{d}{3}x^3 + \frac{e}{2} x^2 + fx + c
\end{equation*}
```

### Reverse chain rule 
Before we define the integral of common functions, it is helpful to consider the **reverse chain rule**. In Topic 12, recall we defined that 
\begin{equation*} \frac{dy}{dx} = \frac{dy}{du} \times \frac{du}{dx}. \end{equation*}
For example, 
\begin{equation*} \\[1em] y = \sin(2x)= \sin(u) \text{ such that }u = 2x \newline \\[1em]
\text{therefore } \frac{dy}{dx} = \cos(u) \times (2) \newline \\[1em] 
 = \cos(2x) \times (2) \end{equation*}

Since integration is the opposite of differentiation, when integrating composite functions $f(g(x))$, we need to divide by the differential of $g(x)$ (such that $g(x)$ is a function of the form $ax + b$.  
For example, 
\begin{equation*} \\[1em] \int 2 \cos(2x) \, dx = 2 \sin(2x) \times \frac{1}{\frac{d}{dx}2x} + c \newline \\[1em]
 = 2 \sin(2x) \times \frac{1}{2} + c = \sin(2x) + c. \end{equation*}

We will use this to define the integral of common functions in the next section.   

### Integration of common functions

From standard differentiation rules we saw in the previous two topics we can find how to integrate common functions. The rules will be outlined below, the non-examinable rationale for some rules requires further integration techniques that we will cover in the next topic, but if you have completed maths A-Level then you should be able to follow them already. 

```{prf:axiom} Integrating an exponential formula 
Since we know $\frac{d(e^x)}{dx} = e^x$ this tells us
\begin{equation*} \int e^x \, dx = e^x + c. \end{equation*}

Since (using the chain rule), we know $\frac{d(e^{ax})}{dx} = ae^{ax}$, this tells us
\begin{equation*} \int e^{ax} \, dx = \frac{1}{a}e^{ax} + c. \end{equation*}
---
The general rule for an exponential of base $a > 0$ (such that $a \neq 1$)   
\begin{equation*} \int a^{kx} \, dx = \frac{a^{kx}}{k\ln|a|} + c \end{equation*}
```

```{prf:axiom} Integration involving logs
Recall that $\frac{d(\ln x)}{dx} = \frac{1}{x}$, which implies that 
\begin{equation*} \int \frac{1}{x} \, dx = \ln |x| + c \end{equation*}
---
The general rule for integrating a logarithmic function is
\begin{equation*} \int \log_a(kx) \, dx = x\log_a(kx) - \frac{x}{\ln|a|} + c \end{equation*}
for $kx > 0$ and $a > 0$ and $a \neq 1$.

---

For the integral of $\ln x$ itself:

\begin{equation*}
\int \ln x \, dx = x \ln |x| - x + c
\end{equation*}

```{tip} But why? Non examinable
:class: dropdown
We can not integrate $\ln |x|$ normally since there is no function such that its derivative is $\ln|x|$. Instead, we use **integration by parts**, letting $u = \ln(x)$ and $v' = 1$ we find that 
\begin{equation*} \int uv' \, dx = uv - \int u'v \, dx \newline \\[1em]
\text{therefore, } \int (1)\ln|x| \, dx = x\ln|x| - \int x \times \frac{1}{x} \, dx \newline
= x\ln|x| - \int 1 \, dx = x\ln|x| - x + c \end{equation*}

*Note: if you have not covered integration by parts, once you have covered the next topic, this will make sense*

```
```{prf:axiom} Trigonometric Functions

From trigonometric derivatives, we obtain the following integration pairs:

Since we know $\frac{d(\sin x)}{dx} = \cos x$ this implies $\int \cos x \, dx = \sin x + c.$

Similarly, $\frac{d(\cos x)}{dx} = -\sin x$ means that $\int \sin x \, dx = -\cos x + c$

---

The general rules, which utilise the reverse chain rule  are
\begin{equation*} \int \cos kx \, dx = \frac{1}{n}\sin kx + c, \end{equation*}
\begin{equation*} \int \sin kx \, dx = -\frac{1}{n}\cos kx + c \end{equation*}
and
\begin{equation*} \int \tan kx \, dx = -\frac{1}{k}\ln|\cos(kx)| + c \end{equation*}
```{tip} But why? Non-examinable
:class: dropdown
To see how to integrate $\tan(x)$ we use *integration by substitution*. First we rewrite tangent in terms of $\sin(x)$ and $\cos(x)$ which gives us
\begin{equation*} \int \tan (kx)\,dx=\int \frac{\sin (kx)}{\cos (kx)}\,dx \end{equation*}
The we $u = \cos(kx)$. Differentiating both sides gives 
\begin{equation*} \frac{du}{dx} = -k\sin(kx) \implies dx = -\frac{1}{k\sin(kx)} \,du \end{equation*}
which we then substitute into our original integral (replacing $\cos(kx)$ with $u$ and $dx$ with $-\frac{1}{k\sin(kx)} \, du$ giving: 

\begin{equation*} \int \frac{\sin(kx)}{\cos(kx)} \, dx = \int \frac{\sin(kx)}{u} \times \frac{-1}{k\sin(kx)} \, du \newline \\[1em] = -\frac{1}{k} \int \frac{1}{u} du =-\frac{1}{k}\ln |{}u|{}+c.\end{equation*}
Then we substitute back in $u = \cos(kx)$ which gives
\begin{equation*} -\frac{1}{k}\ln |{}\cos (kx)|{}+c. \end{equation*}


```


```{seealso} Example

Integrate the expression:

\begin{equation*}
\frac{dy}{dx} = \frac{1}{x} + e^{2x} - \cos 2x
\end{equation*}

```{tip} Click here for solutions 
:class: dropdown
First, we write the integral as a sum of integrals:

\begin{equation*}
y = \int \left(\frac{1}{x} + e^{2x} - \cos 2x\right) \, dx = \int \frac{1}{x} \, dx + \int e^{2x} \, dx - \int \cos 2x \, dx
\end{equation*}

Then we apply the standard integration formulas giving

\begin{equation*}
y = \ln x + \frac{1}{2}e^{2x} - \frac{1}{2}\sin 2x + c
\end{equation*}
```
