# Topic 1: Significant figures and concentration calculations


## Significant figures 
### Chemistry motivation
Decimal places and significant figures are different. We used significant figures to give an answer to an appropriate level of precision. They are important when performing mathematical operations with measured data. 

### Mathematical theory
When quoting a numerical value you must use the correct number of significant figures (sf). For example, the numbers 28.1 and 0.362 are both quoted to 3 sf.

Leading zeros are never counted, but anything between nonzero digits is. The following are all to 3 sf:
\begin{equation*} 0.0000000\textcolor{blue}{507}, 0.\textcolor{blue}{507}, \textcolor{blue}{50.7}  \end{equation*}

As you can see, the trick is to not start counting the significant figures until you have reached your first nonzero digit.

```{tip} Important
When performing a calculation, you can only quote the lowest
number of sf in your final answer:
\begin{equation*} \text{e.g.  } 3.141593 \times 2.71 = 8.51371703 \end{equation*}
\begin{equation*} \text{(7 sf)   } \; \; \; \; \text{    (3 sf)  }\; \; \; \; \; \; \; \; \; \; \; \; \; \end{equation*}
Despite your calculator giving you the answer to 8 decimal places, you can only quote final answer to 3 sf which is 8.51.
```
When working out an answer to a certain number of significant figures, your answer may also require rounding up or down. For example, a temperature of $37.678^{\circ}$C quoted to 3 sf is $37.7^{\circ}$C.

```{seealso} Example
You weigh 3.5 g of analyte to prepare a 250 mL standard solution. The analyte has a relative molecular mass of 122.123 g mol$^{-1}$.
 Calculate the number of moles of analyte in the solution and report your answer to the appropriate number of significant figures.

```{tip} Click here to see the solution
:class: dropdown
Recall that: 
\begin{equation*} \text{number of moles } (n) = \frac{\text{mass of substance }(m)}{\text{mass of 1 mole }(M)} \end{equation*}

\begin{equation*}
\\[1em]
= \frac{3.5}{122.123} \end{equation*}
\begin{equation*} = 0.0286596 \text{mol (to 6 sf)} \end{equation*}
However, as the number of grams is only reported to 2 sf, we can only report
moles to 2 sf = 0.029 mol because the number of significant figures implies accuracy. You cannot report a more accurate value for moles than for the corresponding number of grams.
```

## Concentration and dilution

### Chemistry motivation
Concentration is defined as the amount of solute present or dissolved in a given amount of solvent resulting in a solution. Solute can be a gas (e.g HCl), solid (NaCl) or liquid (AcOH). Solvent is usually a liquid and a common inorganic solvent is water. Common organic solvents are ethanol ($\ce{C2H5OH}$),
ethyl acetate ($\ce{C2H5OCOCH3}$) etc. 

Chemistry often requires making lots of solutions! This often involves dilution of known concentration/s. When making dilutions, \textbf{the number of moles of solute remains constant}. 

### Mathematical theory
There are several ways of representing concentration e.g mol dm$^{-3}$ (or M), g dm$^{-3}$, ppm. The SI units for concentration are mol m$^{-3}$. However, chemists always express concentrations in mol dm$^{-3}$. 


```{prf:axiom} Calculating concentration
To calcuate concentration we use the following formula:

\begin{equation*} \text{Concentraction (C)} = \frac{\text{number of moles (n)}}{\text{volume (V)}} \end{equation*}
```
```{seealso} Example
You weigh 3.5 g of analyte to prepare a 250 mL standard solution. The analyte has a relative molecular mass of 122.123 g mol$^{-1}$.
 Calculate the concentration of the resulting solution in mol dm$^{-3}$ and report your answer to the appropriate number of significant figures.

```{tip} Click here to see the solution
:class: dropdown
Since we know $C = \frac{n}{V}$, we first need to calculate the number of moles, $n$, as we did in the example above. 

\begin{equation*}
\\[1em]
n = \frac{3.5\text{g}}{122.123\text{g mol}^{-1}} = 0.029 \text{ mol } \text{(2 sf)} \end{equation*}

Next, we need to convert the unis of volume to SI units (1cm$^3$ = 0.001 dm$^3$), which gives
\begin{equation*}
\\[1em]
V = 250 \text{cm}^3 \times 0.001 = 0.25 \text{dm}^3 \end{equation*}
and therefore
\begin{equation*}
\\[1em]

C = \frac{0.029 \text{mol}}{0.250 \text{dm}^3} = 0.116 \text{mol dm}^{-3} \text{3 sf}. 
\end{equation*}
However, recall we can only report to the same accuracy as for the number of grams, which gives 0.12 mol dm$^{-3}$ (2 sf).
```

When we make dilutions, the volume (and subsequently the contentration) changes, but the number of moles remains the same, it does not change if we add more solvent.

![](dilution.png)


We can use the face that the moles stay constant in our calulations, and since we know that $C = \frac{n}{V}$, then we can rearrange to obtain $n = CV$. For two solutions, if we know the number of moles in solution 1 ($n_1$) is equal to the number of moles in solution 2 ($n_2$), then this tells us that $C_1V_1=C_2V_2$. 

```{seealso} Example
80 cm$^3$ of water is added to 250 cm$^3$ of 0.90 mol dm$^{−3}$ solution of KOH. Calculate the concentration of the diluted solution.
```{tip} Click here to see the solution
:class: dropdown
Since we know the number of moles n remains the same,
\begin{equation*} n_1= n_2 \implies \end{equation*}
\begin{equation*} C_1V_1= C_2V_2 \end{equation*}
\begin{equation*} C_2 = \frac{C_1V_1}{V_2} = \frac{0.90 \times 0.25}{0.33} = 0.6818 = 0.69 \text{mol dm}^{-3} \text{ (2 sf)} \end{equation*}

Note that in the calculations above, we converted the volumes to the appropriate units as follows:
\begin{equation*} 
\\[1em]
V_1= 250 \text{cm}^3= 250 \times 0.1^3 = 0.25 \text{dm}^3 \end{equation*}
\begin{equation*} 
\\[1em]
V_2= 250 \text{cm}^3 + 80 \text{cm}^3 = 330 \text{cm}^{3} = 0.33 \text{dm}^3 \end{equation*}
```
---
```{tip} Exercises
1. Give the values of (a) 2.554455, (b) 1.72320508, (c) $\pi$ and (d) e to: <br>i. 5, 4 and 3 decimal places, by a single act of rounding in each case,<br>
    ii. 3 significant figures, <br>
using $\pi$ = 3.141592653 and e = 2.718281828.
---
2. You weigh 4.25 g of analyte to prepare a 100 mL standard solution. The analyte has a relative molecular mass of 58.44 g mol$^{-1}$. Calculate the number of moles of analyte in the solution and report your answer to the appropriate number of significant figures.
---
3. A student prepares a primary standard solution for an acid-base titration by weighing out an exact mass of anhydrous sodium carbonate ($\text{Na}_2\text{CO}_3$, relative molecular mass = 105.99~g~mol$^{-1}$). Using an analytical balance, they record a mass of 1.3250~g of the solid, transfer it quantitatively to a 250.0~mL volumetric flask, and dilute it to the calibration mark with deionised water. Calculate the total number of moles of sodium carbonate contained in this standard solution and report the final value to the correct number of significant figures.
---
4. A laboratory technician needs to prepare a weaker alkaline bench reagent. They use a glass pipette to measure out 250 $\text{cm}^3$ of a stock 0.90 $\text{mol dm}^{-3}$ potassium hydroxide ($\text{KOH}$) solution and transfer it into a clean beaker. They then use a burette to add exactly 80 $\text{cm}^3$ of distilled water to the beaker. Calculate the final concentration of the diluted $\text{KOH}$ solution in $\text{mol dm}^{-3}$, reporting your answer to the correct number of significant figures.


```{tip} Click here for solutions

1. <br> (a)(i) We obtain 2.55446 to 5 dp and 2.5545 to 4 dp <br>
&nbsp;&nbsp;&nbsp;&nbsp; (ii) 2.55 to 3 sf <br>
(b)(i) 1.72321 to 5 dp, 1.7232 to 4 dp, 1.723 to 3 dp <br>
&nbsp;&nbsp;&nbsp;&nbsp; (ii) 1.72 to 3 sf <br>
(c)(i) 3.14159 to 5 dp, 3.1416 to 4 dp, 3.143 to 3 dp <br>
 &nbsp;&nbsp;&nbsp;&nbsp; (ii) 3.14 to 3 sf <br>
(d)(i) 2.71828 to 5 dp, 2.7183 to 4 dp, 2.718 to 3 dp <br>
 &nbsp;&nbsp;&nbsp;&nbsp; (ii) 2.72 to 3 sf
---
2. The mass of the analyte is 4.25 g (3 sf). The relative molecular mass is 58.44 g~mol$^{-1}$ (4 sf). Note that the volume of 100 mL is just extra information in this question since we want the total moles in the solution, not the final concentration. Now we know that:
    \begin{equation*} \\[1em] \text{number of moles} = \frac{4.25\text{ g}}{58.44\text{ g mol}^{-1}} = 0.07272416~\text{mol} =  \end{equation*} 
    \begin{equation*}0.0727~\text{mol to 3 sf}. \end{equation*} 
---
3. Using the formula $n = \frac{m}{M}$ and substituting our laboratory measurements gives $n = \frac{1.3250\text{ g}}{105.99\text{ g mol}^{-1}} =  0.012501179 \text{ mol}.$ Reporting this answer to the appropriate number of significant figures gives 0.012501 (or $1.2501 \times 10^{-2}$ mol).
---
4. To find the final concentration, we first determine the total final volume of the mixture by adding the volume of the stock solution to the volume of the added water, which gives $250\text{ cm}^3 + 80\text{ cm}^3 = 330\text{ cm}^3$. We then apply the standard chemical dilution law, $C_1 V_1 = C_2 V_2$, where $C_1$ is the initial concentration ($0.90\text{ mol dm}^{-3}$), $V_1$ is the initial volume ($250\text{ cm}^3$), and $V_2$ is the total final volume ($330\text{ cm}^3$). Rearranging this formula to solve for the final concentration gives 
    \begin{equation*} C_2 = \frac{C_1 V_1}{V_2} = \frac{0.90\text{ mol~dm}^{-3} \times 250\text{ cm}^3}{330\text{ cm}^3} \end{equation*}, which calculates to a raw value of approximately 0.681818~$\text{mol dm}^{-3}$. Looking back at our given values, the initial concentration of 0.90~$\text{mol dm}^{-3}$ limits our precision to two significant figures, so we round our final answer to match. This gives a final concentration of 0.68~$\text{mol dm}^{-3}$.


```
