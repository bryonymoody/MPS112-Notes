### BIDMAS/BODMAS
When carrying out calculations or rearranging expressions you must follow the rules for stardard order of priority. 

The steps are as follows:
- Simplify all operations inside **B**rackets
- Simplify all exponents (**O**rder or **I**ndices) working from left to right
- Perform all **D**ivisions and then **M**ultiplications working from left to right.
- Finally perform all **A**dditions and then **S**ubtractions working from left to right

```{seealso} Examples
First, let us start off with some easier examples, such as,
\begin{equation*} 3^2 + 8 \div 2 - 12 \times 2 = 9 +  8 \div 2 - 12 = 9 + 4 -24 = -11 \nonumber \end{equation*}
and 
\begin{equation*} (a + 2b) \times a - ab = a^2 + 2ab - ab = a^2 +ab \end{equation*}

For some more challenging examples, first we have 
\begin{equation*} \frac{1}{25}(3 \times 10^2)^2 + 3200 \div 4 - 6 \times 120 = \end{equation*}
\begin{equation*} \frac{1}{25} \times (300 \times 300) + 800 - 720 = 3600 + 80 = 3680
\end{equation*}

We can also look at a more complicated algebraic example. 
\begin{equation*} (\theta - 3\theta) \times \phi + \phi \times (\phi\theta)^{-1}-\theta^2\phi^0 \div \theta^3 + \phi^{-1}\theta \phi^2 \times 2\end{equation*}
First we address the brackets and then the indicies giving
\begin{equation*} -2\theta \times \phi + \phi \times \phi^{-1}\theta^{-1} - \theta^2 \div \theta^3 +  \phi\theta \times 2, \end{equation*}
Then we carry out any division and multiplication (in that order) giving  
\begin{equation*} -2\theta\phi + \theta^{-1}-\theta^{-1} +  2\phi\theta \end{equation*}
Finally we find that the whole expression simplifies to 0! 

```
### Fractions
Fractions are simply a different way of showing division, i.e. 

\begin{equation*} a \div b = \frac{a}{b}\end{equation*}

such that $a$ is called the numerator and $b$ is the denominator.

#### Operations with fractions
While it is unlikely you'll have to do fraction operations without a calculator in your day to day life as a chemist, it is worth being aware of how they are carried out. 

##### Addition and subtraction
To add fractions, you must make sure the denominator is the same before you add them together.

```{seealso} Example
Suppose we are given
\begin{equation*}  \frac{1}{4} + \frac{2}{4},\end{equation*}
then we don't need to do anything, we simply need to add the numerators and keep the denominator as it is, which gives
\begin{equation*}  \frac{1}{4} + \frac{2}{4} = \frac{3}{4}.\end{equation*}

It gets slightly mmore complicated if we have different denominators, such as
\begin{equation*} \frac{1}{3} + \frac{1}{4}.\end{equation*}
To approach these questions, we cross multiply which consists of multiplying **both** the numerator and denominator of each fraction by the denominator of the other fraction. For example,
\begin{equation*} \frac{1}{3} = \frac{1}{3} \times \frac{4}{4} = \frac{4}{12}\end{equation*}
It's just a fancy way of multiplying by 1! If we do it to our remaining fraction,
\begin{equation*} \frac{1}{4} = \frac{1}{4} \times \frac{3}{3} = \frac{3}{12}\end{equation*}
then
\begin{equation*} \frac{1}{3} + \frac{1}{4}.\end{equation*}
becomes 
\begin{equation*} \frac{4}{12} + \frac{3}{12} = \frac{7}{12}.\end{equation*}


The same rules apply for subtraction, so we find
\begin{equation*} \frac{1}{3} - \frac{1}{4} = \frac{4}{12} - \frac{3}{12} = \frac{1}{12} \end{equation*}

```
##### Multiplication and division
Thankfully, multiplication is much easier, you simply multiple the numerators of each fraction to get your new numerator and then repeat for the denominator

```{seealso} Example
Despite the denomiators being different, we simply need to do:
\begin{equation*} \frac{2}{3} \times \frac{5}{6} = \frac{10}{18} = \frac{5}{9} \end{equation*}

Remember to always simplify your fraction by diving both the numerator and denominator their largest common factor, in this case it was 2. But do not worry, if you are doing this on your calculator, it will do this step for you.  
```
For division, we require an additional step. You many recall it being referred to as "keep, change, flip" when studying GCSE maths, where we keep the fraction on the left the same, change the division symbol to a multiplication symbol and then flip the fraction on the right. From here you carry out mulitplication of the fractions as in the example above. 

```{seealso} Example
Taking the same fractions as the example above, but now dividing them, we now have
\begin{equation*} \frac{2}{3} \times \frac{5}{6} = \end{equation*}
\begin{equation*} \frac{2}{3} \div \frac{6}{5} = \end{equation*}
\begin{equation*} \frac{12}{15} = \frac{4}{5} \end{equation*}
Remember again to always simplify your fraction by diving both the numerator and denominator their largest common factor, this time it was 3.   
```


### Standard form/scientific notation
Scientific notation (also referred to as standard form) is used by scientists, mathematicians, and engineers to
represent very large or very small numbers. Using exponential notation, large and small numbers can be written in a way that is easier to read.


A positive number is written in scientific notation if it is written as:
\begin{equation*} a × 10^n\end{equation*}
such as the coefficient $a$ is $1 \leq a < 10$, and the exponent $n$ is an integer.

When a number is written in scientific notation, the exponent (n) tells you if the
term is a large or a small number. A positive exponent indicates a large number and a negative exponent indicates a small number between 0 and 1.
```{seealso} Example
Below are some examples of numbers in standard form

```{warning} Note
$5 \times 10^2$ is **smaller** than $4.3 \times 10^7$, since with positive exponents (also know as indices or powers), the larger the exponent, the larger the number.

However, $5 \times 10^{-2}$ is **larger** than $4.3 \times 10^{-7}$ since with negative exponents, the larger the negative exponent, the smaller the number.


### Factorising quadratics
You will need to have a good understanding of how to factorise quadratics, most notably when calculating concentrations at equilibrium. 

#### The General Form of a Quadratic
A quadratic equation is any equation that can be rearranged into the standard form:
\begin{equation*}ax^2 + bx + c = 0 \end{equation*}
where $a$, $b$, and $c$ are constants, and $a \neq 0$. 

First, let us recap factorising simple quadratics, i.e. when $a = 1$ and the equation looks like:
\begin{equation*} x^2 + bx + c = 0. \end{equation*}
To factorise this into two brackets $(x + p)(x + q) = 0$, you need to find two numbers, $p$ and $q$, that satisfy two conditions:
- They multiply to give the constant term: $p \times q = c$
- They add to give the coefficient of $x$: $p + q = b$

```{seealso} Example 
Factorise: $x^2 + 5x + 6$.
```{seealso} Click here to see the answer
:class: dropdown
We need two numbers that multiply to $6$ and add to $5$. The factors of $6$ are $(1, 6)$ and $(2, 3)$. Since $2 + 3 = 5$, our numbers are $2$ and $3$.
\begin{equation*}x^2 + 5x+6 = (x + 2)(x + 3)  \end{equation*}
You may also be asked to solve for $x$ (find the values of $x$ that satisfy  $x^2 + 5x + 6=0$). The solutions (roots) are $x = -2$ or $x = -3$ since these are the two values that make the quadratic equal 0. 
```
```{seealso} Harder Example
Factorise: $x^2 - 3x - 10$
```{seealso} Click here to see the answer
:class: dropdown
We need two numbers that multiply to $-10$ and add to $-3$. Because they multiply to a negative number, one must be positive and one negative.

The factors of $-10$ that add to $-3$ are $-5$ and $+2$. Therefore our factorised format is 
\begin{equation*}x^2 - 3x - 10 = (x - 5)(x + 2)

Solutions: $x = 5$ or $x = -2$. 

**Note**: In a chemistry context, if $x$ represents a concentration, a negative value like $-2$ is physically impossible, so we would reject it and choose $x = 5$.
```


#### Factorising Harder Quadratics ($a > 1$)
When $a \neq 1$, direct guessing becomes a little more difficult but we can still factorise them.

The steps are as follows:

- Multiply $a$ and $c$ together to find the product $ac$.
- Find two numbers that multiply to $ac$ and add to $b$.
- Split the middle term ($bx$) into two terms using these two numbers.
- Factorise the first two terms and the last two terms separately (grouping).


```{seealso} Example
Factorise: $2x^2 + 7x + 3$.
```{seealso} Click here for solutions
:class: dropdown
Here,we have $a=2, b=7, c=3$, and so $ac = 2 \times 3 = 6$. So we need to find two numbers that multiply to $6$ and add to $7$. These are $6$ and $1$.
 
Following this, we split the middle term $7x$ into $6x + 1x$, giving
    \begin{equation*}2x^2 + 6x + 1x + 3 = 0, \end{equation*}
then factorise the first two terms $2x^2 + 6x \to 2x(x + 3)$ and the last two terms $1x + 3 \to 1(x + 3)$.
    
Now we can rewrite the expression as \begin{equation*}2x(x + 3) + 1(x + 3) \end{equation*} and we see that $(x + 3)$ is a common factor. Factorising this out gives
   \begin{equation*}2x^2 + 7x + 3 = (2x + 1)(x + 3)$ \end{equation*} \end{equation*}

To solve for $x$, we need $2x + 1 = 0 \implies x = -0.5$ or $x + 3 = 0 \implies x = -3$.
```

#### The Quadratic Formula

In chemistry, equilibrium constants are rarely nice integers, they are almost always given in scientific notation (e.g., $1.8 \times 10^{-5}$), so you cannot factorise these expressions by eye. 


```{prf:axiom} The quadratic formula
For any quadratic equation $ax^2 + bx + c = 0$, you can find the solutions easily using the **quadratic formula**:
\begin{equation*}x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a} \end{equation*}
```
The symbol $\pm$ means you will calculate two distinct values: one using $+$ and one using $-$. This will also work for all the examples in the section above. 

```{seealso} Example (Chemistry application)
A weak acid has an initial concentration of $0.100\text{ M}$ and a dissociation constant $K_a = 1.8 \times 10^{-5}$. The equilibrium equation simplifies to:
\begin{equation*}x^2 + (1.8 \times 10^{-5})x - (1.8 \times 10^{-6}) = 0  \end{equation*}
Find the equilibrium concentration of hydrogen ions, $x = [\text{H}^+]$.
```{seealso} Click here for solutions
:class: dropdown
First we identify the coefficients: a = 1, $b = 1.8 \times 10^{-5}$ and $c = -1.8 \times 10^{-6}$. 

Next, we substitute these into the quadratic formula:
    \begin{equation*} x = \frac{-(1.8 \times 10^{-5}) \pm \sqrt{(1.8 \times 10^{-5})^2 - 4(1)(-1.8 \times 10^{-6})}}{2(1)} \end{equation*}

Compute the two possible answers:
\begin{equation*} x_1 = \frac{-1.8 \times 10^{-5} + 2.6833 \times 10^{-3}}{2} = 1.33 \times 10^{-3}\text{ M} \end{equation*}
and 
\begin{equation*}x_2 = \frac{-1.8 \times 10^{-5} - 2.6833 \times 10^{-3}}{2} = -1.35 \times 10^{-3}\text{ M} \end{equation*}

Remember, concentration cannot be negative! Therefore, we reject $x_2$ and find that
\begin{equation*} [\text{H}^+] = 1.33 \times 10^{-3}\text{ M} . \end{equation*}
```



## GCSE chemistry content
The topics below are from the GCSE chemistry syllabus and should be familiar to you, but reminders are provided in the sections below. 

### Yield
The **theoretical yield** of a reaction is the maximum amount of product that could be formed from the given amounts of reactants. The **actual** or **absolute yield** is the amount of product that is actually formed when the reaction is carried out in the laboratory.

**Relative yield** is a measure of the effectiveness of a synthetic procedure. Yields can be measured in number of moles or mass. The **percentage yield** is the ratio of the actual yield to the theoretical yield, expressed
as a percentage. Percentage yields express how large the actual yield of a
reaction is compared to the theoretical yield.

```{seealso} Formulae
\begin{equation*}\text{Relative yield} = \frac{\text{actual yield}}{\text{theoretical yield}} \end{equation*}

\begin{equation*}\text{Percentage yield} = \text{Relative yield} \times 100  \end{equation*}

```

```{seealso} Exercise

\begin{equation*}\ce{CH3OH + CH3COOH −> CH3COOCH3 + H2O} \end{equation*}
Based on the molar masses of the reactants and products, $32$g of $\ce{CH3OH}$ reacts
with $60$g of $\ce{CH3COOH}$ to form $74$g of the ester product.
In a particular reaction, $1.8$g of methanol is mixed with an excess of ethanoic acid.
At the end of the experiment $2.6$g of the ester is extracted. What is the percentage
yield?
```{tip} Click here to see the solution

The actual yield is $2.6$g so using the percentage yield formula (with the calculation for relative yield substituted in) we have:
\begin{equation*}\text{Percentage yield} = \frac{\text{actual yield}}{\text{theoretical yield}} \times 100,  \end{equation*}
by substituting the values from our question in, we get
v\text{Percentage yield} = \frac{2.6}{4.2} × 100= 62. \end{equation*}
Therefore, our percentage yield is $62\%$

```
### Units of energy
Any fundamental physical quantity consists of a numerical value and a unit. Scientists use a single system of units: S.I. (Système International). The system is based on a defined unit for each of seven physical quantities and are known as the SI base units and these are outlined in the table below.

![The seven fundamental base units of the Système International](unit_table.png)

Units of all other quantities can be expressed in terms of the seven base units. For commonly used quantities, such as energy, pressure, etc., a set of units derived from the SI base units, and these, for practical reasons, have been given their own name.

![](derived_units.png)

Energy is defined as the ability to do work e.g the it can be the force required to move an object.
\begin{equation*}\text{Work} = \text{Force} \times \text{displacement} \end{equation*}

To work out the units we do the following
\begin{equation*}\text{Units of energy} = \text{units of force} \times \text{units of distance} \end{equation*}
\begin{equation*}= \text{kg m s}^{-2} \times \text{m}= \text{kg m}^2 \text{s}^{-2}= \text{Joule} \end{equation*}

#### Units of Concentration and Rates of reaction
Concentration can be defined as the amount of solute present or dissolved in a given amount of solvent resulting in a solution. 

\begin{equation*}\text{Units of Concentration} = \frac{\text{units of amount of solute}}{\text{units of volume}} = \frac{\text{no of moles}}{\text{dm}^{-3}}  \end{equation*}
\begin{equation*}= \frac{\text{mol}}{\text{dm}^{-3}} = \text{mol dm}^{-3} \end{equation*}

Rate of reaction is defined as the increase in the concentration of a product per unit time or the decrease in the concentration of a reactant per unit time. 
\begin{equation*}\text{Units of rate} = \frac{\text{units of concentration change}}{\text{units of time}} = \frac{\text{mol dm}^{-3}}{s} = \text{mol dm}^{-3} \text{s}^{-1} \end{equation*}

```{note}
Some important things to note:
- In scientific notation the physical quantity is always be followed by its units.
- If the units are “mixed”, as in metres per second, it should be written as ms$^{-1}$. The
“/” sign **should not be used** for units, always use negative powers.
- When a unit is derived from a person’s name, such as newtons, the first letter of
the unit’s name is given in lower case; however, the unit symbol is given with a
capital letter. e.g, the magnitude of the force is 10 newtons. \begin{equation*} \text{F} = 10 \text{N} \end{equation*}
```

#### Unit prefixes
The SI units are not always suitable to express very large or very small numbers. • e.g SI unit of length, is not convenient for expressing the distance between atoms in a molecule. To be able to do so, we make use of prefixes.
The prefixes are available, in general, in multiples of $10^3$.

![](prefix_table.png)

```{note}
The kilogram is anomalous: although it is a base unit of mass, it already has a prefix. Therefore 1 kg = 1000 g = $10^3$ g. So prefixes are attached to the gram.
\begin{equation*}1\text{g} = \frac{1}{1000} = \frac{1}{10^3} = 10^{-3}\text{ kg} \end{equation*}
\begin{equation*}1\text{mg} = \frac{1}{1000} = 0.001 = 1 \times 10^{-3}\text{ g} = 1^{-3}\text{ g} \end{equation*}
```
#### Converting units
Different units are linked by a conversion factor. e.g. 1cm = 10mm, so the conversion factor between these units is 10. When converting units, think about the result in the new units (should it be bigger or smaller?). For example, if we want to convert from cm to mm; do we multiply or divide by 10? Since we know that the mm is smaller than the cm, we should get a bigger number.
\begin{equation*}\text{Therefore: }3\text{ cm} = 3 \times 10 = 30\text{ mm} \end{equation*}

Conversely, to convert from mm to cm, we should get a smaller number.
\begin{equation*}\text{Therefore: }30\text{ mm} = 30 \div 10 = 3\text{ cm} \end{equation*}

#### Powers of units
Volumes can commonly be expressed in litres (L), m$^{3}$, dm$^{3}$ or cm$^{3}$. S.I units for volume is m$^3$ 
\begin{equation*}1\text{ dm}^3 = 1000\text{ mL} \end{equation*}
\begin{equation*}1\text{ L} = 1\text{ dm}^3 \end{equation*}

Powers of units apply to the prefix as well as the unit they modify. In other words, the d in front of dm$^{-3}$ means 0.1m, not 0.1m$^{3}$. When we use dm$^3$ we mean (dm)$^3$.
\begin{equation*}1 \text{ dm}= 0.1\text{ m} \end{equation*}
\begin{equation*}1 \text{ dm}^3 = (0.1 \text{ m})^3 = (0.1)^3(\text{ m})^3 = (0.1 \times 0.1 \times 0.1)\text{ m}^3 = 0.001\text{ m}^3 \end{equation*}
\begin{equation*}1 \text{ dm}^3= (10^{-1}\text{ m})= (10^{-1})^3(\text{ m})^3= 10^{-3}\text{ m}^3 \end{equation*}

### Moles
The mole (or mol) is the SI unit for amount of substance, such that 1 mole of any substance contains $6.02214076 \times 10^{23}$ particles (to 9 sf). This number is known as Avogadro’s constant, $N_a$. In practice, 4 significant figures is an appropriate level of accuracy, $6.022 \times 10^{23}.$

```{note}
Do not worry if you do not feel confident with significant figures, this will be revised during the lectures.
```


The historical definition of the mole is based on the amount of substance corresponding to the number of atoms in exactly 12 grams of $\ce{^12C}$. Thus, 1 mole of $\ce{^12C}$ has a mass of exactly 12g and contains $6.022 \times 10^{23}$ atoms of carbon.

```{seealso} Example
To obtain 1 mole of a substance, you need the mass (g) of the substance to match the relative molecular mass (RMM)/relative atomic mass (RAM).

1 mole of $\ce{H2O}$ has a mass of exactly 18g and contains $6.022 \times 10^{23}$ molecules of $\ce{H2O}$ since the RMM of $\ce{H2O}$ = 18g mol$^{-1}$.

Similarly, 1 mole of sodium metal has a mass of 23 g and contains $6.022 \times 10^{23}$ atoms of Na since the RAM of Na = 23g mol$^{-1}$.
```

As chemists we are interested in the ratio of molecules involved in a reaction (i.e. the stoichiometry). 

For example, 1 molecule of benzophenone reacts with 1 molecule of sodium borohydride, producing 1 molecule of alcohol product

![insert title](chem_form.png)

However, we don’t carry out our reaction on the molecular scale, one molecule at a time! In the lab, we typically use a gram scale. Since molecules are so tiny, this corresponds to a huge number of individual molecules. Hence, the mole (a very large number) is used as a scaling factor so we can talk about the ratios of reacting molecules using convenient, relatable values.

```{seealso} Exercise
You weigh 3.5g of analyte to prepare a 250mL standard solution. The analyte has a relative molecular mass of 122.123g mol$^{-1}$. Calculate the number of moles of analyte in the solution.


```{tip} Click here to see the solution
:class: dropdown
Given that 
\begin{equation*}\text{Number of moles} = \frac{\text{mass of substance}}{\text{mass of 1 mole}} \end{equation*}
\newline
= \frac{3.5}{122.123}
\newline
= 0.0286596 \text{ mol (to 6 sf)}. \end{equation*}

However, as the number of grams is only reported to 2 sf, we can only report moles to 2 sf which gives 0.029 mol. The number of significant figures implies accuracy. You cannot report a more accurate value for moles than for the corresponding number of grams.

```
