# Topic 10: Argand diagrams and polar coordinates

## The Argand Diagram

Complex numbers are visualized on a Cartesian plane by mapping the real part to the x-axis and the imaginary part to the y-axis. The complex number $z = a + bi$ is represented geometrically as a point $(a, b)$, with the 
horizontal axis corresponding to the real component of the complex number and the vertical axis corresponding to the imaginary axis.

Suppose we wanted to represent the imaginary number $z = 3 + 4i$ on an argand diagram. This equates to plotting the point (3,4).

![](argand.png)


## The Modulus and Argument of $z$

The distance from the origin to the point $(a, b)$ is defined as the **modulus** ($r$ or $\vert{}z\vert{}$) of the complex number $z = a + bi$ and can be calculated using Pythagoras' theorem.  

\begin{equation*} \vert{}z\vert{} = r = \sqrt{a^2 + b^2} \end{equation*}

Whereas the **argument** ($\theta$ or $\arg(z)$) is the angle measured counterclockwise from the positive real axis to the line segment connecting the origin to $(a, b)$ (when $z$ is in the first quadrant in the argand diagram) and we use trigonometic ratios to work it out, specifically:

\begin{equation*} \theta = \tan^{-1} \left(\frac{b}{a}\right) \end{equation*}

![](modarg.png)

For the complex number $z = 3 + 4i$ we calculate our modulus ($\vert{}z\vert{}$) as follows:
\begin{equation*}
|z| = \sqrt{a^2 + b^2} = \sqrt{3^2 + 4^2} = \sqrt{9 + 16} = \sqrt{25} = 5
\end{equation*}

The argument ($\arg(z)$) is  
\begin{equation*}
\arg(z) = \theta = \tan^{-1}\left(\frac{b}{a}\right) = \tan^{-1}\left(\frac{4}{3}\right) \approx 0.9273 \text{ rad} \quad (\approx 53.13^\circ)
\end{equation*}

Note that we defined the argument above $z$ in the first quadrant on an argand diagram. For complex numbers in other quadrants we take the postive of $a$ and $b$ and use the resulting argument $\theta'$ to calculate:
- $\theta = \pi - \theta'$ (second quadrant)
- $\theta = -\pi + \theta'$ (third quadrant)
- $\theta = -\theta'$ (fourth quadrant)

Notice that the argument is conventionally taken to be in the range $-\pi < \theta \le \pi$. We look at some examples of this now.

```

### The Polar Form for Complex Numbers

```{prf:axiom} Polar form
The complex number $z=a+bi$ can be written in polar form:
\begin{equation*} z = r(\cos\theta + i\sin\theta) \end{equation*} 
such that $r$ is the modulus of $z$ and $\theta$ is the argument. 
``` 
To see why the polar form $z = r(\cos\theta + i\sin\theta)$ is the same as $z = a + bi$ we use standard trigonometry. 

Looking at the right-angled triangle formed in the Argand diagram above we have the following components:
- Hypotenuse: $r$ (the magnitude or modulus, $\vert{}z\vert{}$)
- Adjacent side: $a$ (the real part)
- Opposite side: $b$ (the imaginary part)
- Angle: $\theta$ (the argument, $\text{arg}(z)$)

Using standard trigonometry ($\text{SOH CAH TOA}$):
\begin{equation*}
\cos(\theta) = \frac{\text{Adjacent}}{\text{Hypotenuse}} = \frac{a}{r} \implies a = r\cos(\theta) \end{equation*}

\begin{equation*} \sin(\theta) = \frac{\text{Opposite}}{\text{Hypotenuse}} = \frac{b}{r} \implies b = r\sin(\theta)\end{equation*}


Therefore we can infer that 
\begin{equation*} 
a + bi = r\cos(\theta) + ri\sin(\theta) = r(\cos(\theta) + i\sin(\theta))
\end{equation*}

### Complex form of complex equations
There is, in fact, a third way of representing complex numbers which is the **exponential form**

```{prf:axiom} Exponential form of complex numbers
The complex number $z=a+bi$ can be written in polar form:
\begin{equation*} z = r e^{i\theta}
\end{equation*} 
such that $r$ is the modulus of $z$ and $\theta$ is is the argument. 
``` 

You might wonder: why bother to express the same thing in three different ways? However, depending on what you wish to do with your complex numbers, one of the three formats might be preferable. To demonstrate this, let us return to our example from above, $z = 3 + 4i$. We can represent this in the following formats:

\begin{equation*}
z = 3 + 4i = 5(\cos(0.9273) + i\sin(0.9273)) = 5e^{0.9273i} \newline
\\[1em]
\text{Cartesian }(x + iy) = \text{Polar }(r(\cos\theta + i\sin\theta)) = \text{Exponential }(re^{i\theta}) 
\end{equation*}

In the cartesian format, it is very easy to read off the real (3) and imaginary ($4i$) components, and carrying out addition and subtraction with two complex numbers is fairly simple. However multiplication and (especially) division are more difficult.

In the exponential format, multiplication and division become much easier. We simply use laws of indices to work it out. 

```{seealso} Example
Suppose we have $z_1 = 6 + 4i$ and $z_2 = 5 + 2i$. Before, we calculated $z_1 \div z_2$ in the form $a + bi$. Now, convert $z_1$ and $z_2$ into the exponential form instead and then work out $z_1$ and $z_2$.
```{tip} Click here for solutions
:class: dropdown
\begin{equation*}
z_1 = (\sqrt{4^2 + 6^2})e^{i\tan^{-1}(4/6)} = 7.21e^{0.588i}
\end{equation*}
and 
\begin{equation*}
z_2 = (\sqrt{2^2 + 5^2})e^{i\tan^{-1}(2/5)} = \sqrt{29}e^{0.381i}
\end{equation*}

Now we have both equations in exponential format, we can use the fact that 
\begin{equation*}
r_1 e^{k_1} \div r_2 e^{k_2} = \frac{r_1 e^{k_1}}{r_2 e^{k_2}} = \frac{r_1}{r_2} \times \frac{e^{k_1}}{e^{k_2}} = \frac{r_1}{r_2}e^{k_1 - k_2}
\end{equation*}

Now, looking back to our example, this means that
\begin{equation*}
z_1/z_2 = \frac{7.21}{\sqrt{29}}e^{(0.588-0.38)i} = 1.34e^{0.208i}
\end{equation*}

```
### Euler's formula
Given 
\begin{equation*}
z = re^{i\theta} = r(cos(\theta) + i sin(\theta)),
\end{equation*}
we can cancel our r giving
\begin{equation*}
e^{i\theta} = cos(\theta) + i sin(\theta).
\end{equation*}
and this relationship is known as **Eulers Formula**

Now, suppose that $\theta = \pi$. We find that 
\begin{equation*}
e^{i\pi} = cos(\pi) + i sin(\pi)
\newline e^{i\pi} = -1 + i (0)
\end{equation*}
which rearranges to give
\begin{equation*}
e^{i\pi} + 1 = 0
\end{equation*}

which is known as **Eulers Identity** which is one of the most famous identities in mathematics.

### Powers of complex numbers

The interesting thing about Eulers formula is that is makes thinking about powers of complex numbers considerably easier. 

Given $z = re^{i\theta}$, we can deduce that 
\begin{equation*}
z^n = (re^{i\theta})^n = r^n(e^{i\theta})^n = r^ne^{in\theta}
\end{equation*}
as a polar coordinate this is the same as saying

\begin{equation*}
z^n = r^ne^{in\theta} = r^n(cos(n\theta)+isin(n\theta))
\end{equation*}

```{seealso} Example - Powers of complex numbers
Suppose we have the complex number $z = 5 - 6i$. Express $z^6 = (5 - 6i)^6$ in the form $a + bi$.

```{tip} Click here for solutions
:class: dropdown
First, let us find the modulus ($r$) and argument ($\theta$) of $z$ so that we can express it in polar coordinates: 
\begin{equation*}
r = \sqrt{5^2 + (-6)^2} = \sqrt{25 + 36} = \sqrt{61}
\end{equation*}

Since $a > 0$ and $b < 0$, the complex number $5 - 6i$ lies in the fourth quadrant:
\begin{equation*}
\theta = -\arctan\left(\frac{6}{5}\right) \approx -0.87606 \text{ radians} \quad (\text{or } -50.194^\circ).
\end{equation*}

Therefore, the polar form of $z$ is:
\begin{equation*}
z = \sqrt{61} \left(\cos(-0.87606) + i \sin(-0.87606) \right).
\end{equation*}

From this, we can deduce that 
\begin{equation*}
\\[1em]
z^6 = \left(\sqrt{61}\right)^6 \left( \cos(6\theta) + i\sin(6\theta) \right).
\end{equation*}

Evaluating $6\theta$ requires an additional step, since we tend to express arguments of complex numbers in the $[-\pi, \pi]$ range. 

\begin{equation*}
6\theta =  6 \times (-0.876058) = -5.25635 \text{ radians} \quad (\text{or } -301.166^\circ)
\end{equation*}

then we add $2\pi$ to get us in the desired range

\begin{equation*}
6\theta + 2\pi \approx -5.25635 + 6.28319 = 1.02684 \text{ radians} \quad (\text{or } 58.834^\circ)
\end{equation*}

So finally we have our result in polar form:
\begin{equation*}
z^6 = 226,981 \left( \cos(1.02684) + i\sin(1.02684) \right).
\end{equation*}

Converting this back to Cartesian format we use the following
- $\cos(1.02684) \approx 0.517514$
- $\sin(1.02684) \approx 0.855675$

Then substituting these back into the expression we find
\begin{equation*}
\text{Re}(z^6) = 226,981 \cdot 0.517514 = 117,466
\end{equation*}

\begin{equation*}
\text{Im}(z^6) = 226,981 \cdot 0.855675 = 194,222
\end{equation*}
And so we deduce that $(5 - 6i)^6$ is:
\begin{equation*}
z^6 = 117,465 + 194,222i.
\end{equation*}
