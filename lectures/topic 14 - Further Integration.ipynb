## Mathematical theory
To solve more complicated integrals, there are two new strategies that we will employ: integration by substitutions and integration by parts. 

### Integration by substitution
In Topic 13 we looked at how we can reverse the chain rule to help us integrate composite functions (functions of the form $f(g(x))$) but this can be expanded to help us integrate more complex expressions:
```{prf:axiom} Integration by substitution
For functions of the form $f(g(x)) \times g'(x)$ then 
\begin{equation*} \int f(g(x)) \times g'(x) \, dx = \int f(u) \, du \end{equation*}
such that $u = g(x)$. 

In other words, if you have a composite function, multiplied by (some multiple of) the derivative of the function inside the other function, you can simplify the integral greatly. 
```
Now let us look at some examples to see how to carry out this method. 

```{seealso} Example
Evaluate $\int x e^{x^2} \, dx.$
```{tip} Click here to see solutions
:class: dropdown
*Step 1: Choose your substitution $u = g(x)$* <br>
Let us try the substution $u = x^2$. <br>

---

*Step 2: Differentiate your substitution function* <br>
\begin{equation*} \frac{du}{dx} = 2x \end{equation*}

---

*Step 3: Rearrange to find $dx$* <br>
\begin{equation*} \frac{du}{dx} = 2x \implies dx = \frac{1}{2x} \, du  \end{equation*}

---

*Step 4: Substitute $u$ and $dx$ back into your original integral* <br>
Since $ dx = \frac{1}{2x}du$ and $u = x^2$, then 
\begin{equation*} \int x e^{x^2} \, dx = \int xe^u \times \frac{1}{2x}\, du \newline \\[1em] = \frac{1}{2} \int e^u \, du = \frac{1}{2}e^u + c    \end{equation*}

---

*Step 5: Replace $u$ with the original function:* <br>
\begin{equation*}  \frac{1}{2}e^u + c =    \frac{1}{2}e^{x^2} + c  \end{equation*}

```
There are also examples where substitution can help us simplify the integral even if one part of the integral is not the derivative of the other. These are harder to spot, but it gets easier with practice.  
```{seealso} Example
Evaluate 
\begin{equation*} \int \frac{x}{\sqrt{1-x}} \, dx \end{equation*}

```{tip} Click here to see the solution
:class: dropdown
In fact, there are multiple ways to solve this integral, here we will look at two possible substitutions: <br>
**Substitution 1:**
We let $\textcolor{blue}{u = 1 - x}$ and by rearranging for $x$ and differentiating this gives:
\begin{equation*} \textcolor{darkorange}{x = 1 - u} \quad \text{and} \quad \textcolor{red}{dx = -du} \end{equation*}

Then we substitute these expressions into the integral giving
\begin{equation*} \int \frac{\textcolor{darkorange}{x}}{\sqrt{\textcolor{blue}{1-x}}} \, \textcolor{red}{dx} = \int \frac{\textcolor{darkorange}{1 - u}}{\sqrt{\textcolor{blue}{u}}} \textcolor{red}{(-du)} = \int \frac{u - 1}{u^{\frac{1}{2}}} \, du \end{equation*}
The final step before integrating is to split the fraction into individual power terms:
\begin{equation*} \int \left(u^{\frac{1}{2}} - u^{-\frac{1}{2}}\right) du. \end{equation*}

Now we can integrate term by term giving 
\begin{equation*} \frac{u^{\frac{3}{2}}}{\frac{3}{2}} - \frac{u^{\frac{1}{2}}}{\frac{1}{2}} + c = \frac{2}{3}u^{\frac{3}{2}} - 2u^{\frac{1}{2}} + c \end{equation*}

and factor out $\frac{2}{3}u^{\frac{1}{2}}$ to simplify giving
\begin{equation*} \\[1em] \frac{2}{3}u^{\frac{1}{2}}(u - 3) + c \end{equation*}
and the final step is to substitute $\textcolor{blue}{u = 1 - x}$ back into the expression:
\begin{equation*} \\[1em] \frac{2}{3}\textcolor{blue}{u}^{\frac{3}{2}} - 2\textcolor{blue}{u}^{\frac{1}{2}} + c = \frac{2}{3}\sqrt{\textcolor{blue}{1 - x}}\big((\textcolor{blue}{1 - x}) - 3\big) + c = \newline \\[1em] -\frac{2}{3}\sqrt{1 - x}(x + 2) + c \end{equation*}

---

**Substitution 2:** <br>
To evaluate the integral using $\textcolor{blue}{u = \sqrt{1-x}}$, we square both sides first to solve for $x$:
\begin{equation*} \\[1em] u^2 = 1 - x \implies \textcolor{darkorange}{x = 1 - u^2} \end{equation*} 

Then differentiating with respect to $u$ gives 
\begin{equation*} \frac{dx}{du} = -2u \implies \textcolor{red}{dx = -2u \, du} \end{equation*}

Replacing $\textcolor{darkorange}{x}$, $\textcolor{blue}{\sqrt{1-x}}$, and $\textcolor{red}{dx}$ in the integral we find
\begin{equation*} \\[1em] \int \frac{\textcolor{darkorange}{x}}{\textcolor{blue}{\sqrt{1-x}}} \, \textcolor{red}{dx} = \int \frac{\textcolor{darkorange}{1 - u^2}}{\textcolor{blue}{u}} \textcolor{red}{(-2u) \, du} \end{equation*}

We then cancel $u$ and expand the remaining expression to make things simpler, giving
\begin{equation*} \int -2(1 - u^2) \, du = \int (2u^2 - 2) \, du \end{equation*}

Now we integrate term by term 
\begin{equation*} \\[1em]\frac{2}{3}u^3 - 2u + c = \frac{2}{3}u(u^2 - 3) + c \end{equation*}

To finish off, we substitute $\textcolor{blue}{u = \sqrt{1 - x}}$ and $u^2 = 1 - x$ back into the expression to get 
\begin{equation*} \\[1em] \frac{2}{3}\textcolor{blue}{u}(u^2 - 3) + c \ = \frac{2}{3}\textcolor{blue}{\sqrt{1 - x}}\big((1 - x) - 3\big) + c \newline \\[1em] = -\frac{2}{3}(x + 2)\sqrt{1 - x} + c \end{equation*}
---
*Note: in these examples, we had the additional step of writing $x$ in terms of some function of $u$. This is needed for the problems where replacing $dx$ does not cancel out any remaining occurences of $x$ in the integral*

```

### Integration by parts
When differentiating the product of two functions ($f(x) \times g(x)$) we used the **product rule**. When integrating the product of two functions we use **integration by parts**. 

```{prf:axiom} Integration by parts
Given a function $f(x)g(x)$ such that we define $f(x) = u $ and $g(x) = \frac{dv}{dx}$
\begin{equation*} \int f(x)g(x) \, dx = \int u \, \frac{dv}{dx} \, dx = uv - \int v \, \frac{du}{dx} \, dx \end{equation*}
```
The difficulty with this technique is that you need to choose which function will be equal to $u$ and which will be equal to $\frac{dv}{dx}$. Since we will differentiate $u$, you typically choose the function that simplifies the most after differentiating to be $u$. 

```{seealso} Example
Integrate $x\cos(x)$ with respect to $x$. 
```{tip} Click here to see solutions
:class: dropdown
The two functions are $x$ and $\cos(x)$. Since $x$ simplifies to 1 when we differentiate it, whereas $\cos(x)$ differentiates to $- \sin(x)$ we determine that $x$ simplifies the most when differentiating and set it to be $u$. Therefore
\begin{equation*} u = x \text{ and } \frac{dv}{dx} = \cos(x). \end{equation*} 
Next we find 
\begin{equation*} \frac{du}{dx} = \frac{d}{dx}x =  1 \end{equation*}
and 
\begin{equation*} v = \int \frac{dv}{dx} \, dx = \int \cos(x) \, dx = \sin(x) \end{equation*}
Substituting this into 
\begin{equation*} \int u \, \frac{dv}{dx} \, dx = uv - \int v \, \frac{du}{dx} \, dx \end{equation*}
 gives
\begin{equation*} \int x \, \cos(x) \, dx = x\sin(x) - \int \sin(x) \, (1) \, dx \newline \\[1em]
= x\sin(x) - (- \cos(x)) + c = x\sin(x) + \cos(x) + c   \newline
\end{equation*}
```
```{seealso} Example - revisiting $\int \frac{x}{\sqrt{1-x}} \, dx$ 
Evaluate 
\begin{equation*} \int \frac{x}{\sqrt{1-x}} \, dx \end{equation*}  
using integration by parts. 
```{tip} Click here to see solutions
:class: dropdown
The two functions are $x$ and $\frac{1}{\sqrt{1-x}} = (1-x)^{-\frac{1}{2}}$. Since $x$ simplifies to 1 when we differentiate it, whereas $(1-x)^{-\frac{1}{2}}$ differentiates to $\frac{1}{2}(1-x)^{-\frac{3}{2}}$, we determine that $x$ simplifies the most when differentiating and set it to be $u$. Therefore
\begin{equation*} u = x \text{ and } \frac{dv}{dx} = (1-x)^{-\frac{1}{2}}. \end{equation*}
Next we find
\begin{equation*} \frac{du}{dx} = \frac{d}{dx}x = 1 \end{equation*}
and using the chain rule we find
\begin{equation*} v = \int \frac{dv}{dx}  dx = \int (1-x)^{-\frac{1}{2}} \, dx = -2(1-x)^{\frac{1}{2}} \end{equation*}
Substituting this into
\begin{equation*} \int u \, \frac{dv}{dx} \, dx = uv - \int v \, \frac{du}{dx} \, dx \end{equation*}
gives
\begin{equation*} \int \frac{x}{\sqrt{1-x}} \, dx = -2x(1-x)^{\frac{1}{2}} - \int -2(1-x)^{\frac{1}{2}} \, (1) \, dx \newline \\[1em]
= -2x(1-x)^{\frac{1}{2}} + 2\int (1-x)^{\frac{1}{2}} \, dx \newline \\[1em]
= -2x(1-x)^{\frac{1}{2}} + 2\left(-\frac{2}{3}(1-x)^{\frac{3}{2}}\right) + c \newline \\[1em]
= -2x(1-x)^{\frac{1}{2}} - \frac{4}{3}(1-x)^{\frac{3}{2}} + c = -\frac{2}{3}(x + 2)\sqrt{1-x} + c \newline
\end{equation*}
as before.
```
